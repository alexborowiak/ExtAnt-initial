```python
def _hist_preprocess_single_member(ds: xr.Dataset) -> xr.Dataset:
    """Subset one file to tasmax over the Dome C box and label it by variant.

    Parameters
    ----------
    ds : xr.Dataset
        A single CMIP6 file as opened by ``open_mfdataset``.

    Returns
    -------
    xr.Dataset
        ``tasmax`` only, on a (member, time, lat, lon) grid, encoding stripped.
    """
    ds = (
        ds[["tasmax"]]
        .sel(lat=slice(-82, -78), lon=slice(120, 129))
        .expand_dims(member=[ds.attrs["variant_label"]])
    )
    ds.encoding.clear()
    ds.tasmax.encoding.clear()
    ds.time.encoding.clear()
    return ds
```

```python
def group_by_member(fnames: list[str]) -> dict[str, list[str]]:
    """Group file paths by the rXiYpZfW member id in their filename."""
    by_member = defaultdict(list)
    for p in fnames:
        by_member[re.search(r"_(r\d+i\d+p\d+f\d+)_", p).group(1)].append(p)
    logger.debug("grouped %d files into %d members", len(fnames), len(by_member))
    return by_member


def get_members_already_run(out: str) -> list[str]:
    """Member ids present in the Zarr store `out`; empty list if `out` does not exist."""
    if not os.path.exists(out):
        logger.debug("no store at %s", out)
        return []
    members = list(xr.open_dataset(out, engine='zarr').member.values)
    logger.debug("found %d members in %s", len(members), out)
    return members


def make_batches(
    by_member: dict[str, list[str]],
    members_already_run: list[str],
    batch_size:int=5
) -> list[list[list[str]]]:
    """File lists for the members of `by_member` not in `members_already_run`, in batches of 5."""
    nested = [sorted(f) for m, f in by_member.items() if m not in members_already_run]
    batches = [nested[i:i + batch_size] for i in range(0, len(nested), batch_size)]
    logger.debug("%d members pending -> %d batches", len(nested), len(batches))
    print(list(map(len, nested)))
    return batches

def group_by_member(fnames: list[str]) -> dict[str, list[str]]:
    """Group file paths by the rXiYpZfW member id in their filename."""
    by_member = defaultdict(list)
    for p in fnames:
        by_member[re.search(r"_(r\d+i\d+p\d+f\d+)_", p).group(1)].append(p)
    logger.debug("grouped %d files into %d members", len(fnames), len(by_member))
    return by_member


def get_members_already_run(out: str) -> list[str]:
    """Member ids present in the Zarr store `out`; empty list if `out` does not exist."""
    if not os.path.exists(out):
        logger.debug("no store at %s", out)
        return []
    members = list(xr.open_dataset(out, engine='zarr').member.values)
    logger.debug("found %d members in %s", len(members), out)
    return members


def make_batches(
    by_member: dict[str, list[str]],
    members_already_run: list[str],
    batch_size:int=5
) -> list[list[list[str]]]:
    """File lists for the members of `by_member` not in `members_already_run`, in batches of 5."""
    nested = [sorted(f) for m, f in by_member.items() if m not in members_already_run]
    batches = [nested[i:i + batch_size] for i in range(0, len(nested), batch_size)]
    logger.debug("%d members pending -> %d batches", len(nested), len(batches))
    print(list(map(len, nested)))
    return batches


def write_group(
    group: list[list[str]],
    out: str,
    preprocess: Callable[[xr.Dataset], xr.Dataset],
    append: bool,
) -> list[str]:
    """Open the file lists in `group` along (member, time) and write them to `out`."""
    ds = xr.open_mfdataset(
        group,
        preprocess=preprocess,
        combine='nested',
        concat_dim=['member', 'time'],
        parallel=True,
        data_vars='minimal',
        coords='minimal',
        compat='override',
        join='override',
        chunks={'time': -1}
    )
    if logger.isEnabledFor(logging.DEBUG): display(ds)
    ds = ds.load()
    ds = ds.chunk({"member": 1, "time": -1, "lat": -1, "lon": -1})
    if append:
        ds.to_zarr(out, append_dim="member")
    else:
        logger.info("creating store %s", out)
        ds.to_zarr(out, mode="w")
    members = list(ds.member.values)
    ds.close()
    return members


def write_batches(
    batches: list[list[list[str]]],
    out: str,
    preprocess: Callable[[xr.Dataset], xr.Dataset],
) -> None:
    """Write each batch to `out`; on failure retry its members one at a time, skipping any that fail."""
    append = os.path.exists(out)
    for i, batch in enumerate(batches):
        logger.info("batch %d/%d: opening %d members", i + 1, len(batches), len(batch))
        logger.debug(batch)
        try:
            members = write_group(batch, out, preprocess, append)
            append = True
            logger.info("wrote batch %d/%d: %s", i + 1, len(batches), members)
        except Exception:
            logger.exception("batch %d/%d failed, retrying members individually", i + 1, len(batches))
            for files in batch:
                try:
                    members = write_group([files], out, preprocess, append)
                    append = True
                    logger.info("wrote member %s", members)
                except Exception:
                    logger.exception("skipping member, files: %s", files)
```

```python
def plot_phase_percent(layers, phases, labels=None, title=None, ylabel='percent', figsize=(8, 7)):
    """Plot one panel per phase, overlaying every layer against time.

    Args:
        layers (list[dict]): Each has 'data' (xr.DataArray with 'phase' and 'time'),
            optional 'mean_line' (bool) to add its time mean as a dotted line, and
            Line2D style kwargs such as 'label' and 'color'.
        phases (array-like): Phase values, one panel each.
        labels (dict | None): Panel titles keyed by phase value; the phase value is
            used where absent.
        title (str | None): Figure title.
        ylabel (str): Y label applied to every panel.
        figsize (tuple[float, float]): Figure size.

    Returns:
        tuple[plt.Figure, np.ndarray]: Figure and axes.
    """
    labels = labels or {}
    fig, axes = plt.subplots(len(phases), 1, figsize=figsize, sharex=True, constrained_layout=True)
    for ax, phase in zip(axes, phases):
        for layer in layers:
            style = {k: v for k, v in layer.items() if k not in ('data', 'mean_line')}
            data = layer['data'].sel(phase=phase, drop=True)
            data.plot.line(ax=ax, **style)
            if layer.get('mean_line'):
                ax.axhline(data.mean('time'), color=style.get('color'), ls=':', lw=1)
        ax.set_title(labels.get(phase, phase), loc='left')
        ax.set_ylabel(ylabel)
        ax.set_xlabel('')
    axes[0].legend(frameon=False, loc='upper left')
    fig.suptitle(title)
    return fig, axes
```

```python
def plot_index_plume(layers, ax, pairs, median=0.5, x='time', dim='member'):
    """Draw a quantile plume across ensemble members for each layer.

    Args:
        layers (list[dict]): Each has 'data' (xr.DataArray with `x` and `dim`) plus
            plot_plume style kwargs such as 'label' and 'color'.
        ax (plt.Axes): Axis to draw on.
        pairs (tuple[tuple[float, float], ...]): (lower, upper) levels, outermost first.
        median (float): Quantile level drawn as a line.
        x (str): Coordinate name for the x axis.
        dim (str): Dimension the quantiles are taken over.
    """
    quantiles = np.unique([*np.ravel(pairs), median])
    for layer in layers:
        style = {k: v for k, v in layer.items() if k != 'data'}
        plot_plume(ax, layer['data'].quantile(quantiles, dim=dim), pairs,
                   median=median, x_dim=x, **style)
```

```python
PHASES = xr.DataArray([-1, 0, 1], dims='phase', coords={'phase': [-1, 0, 1]})


def phase_index(anomaly, threshold):
    """Classify a normalised climate index into phases by magnitude.

    Args:
        anomaly (xr.DataArray): Climate index anomaly.
        threshold (float): Magnitude above which an anomaly counts as an event.

    Returns:
        xr.DataArray: -1 (negative), 0 (neutral), 1 (positive); NaN preserved.
    """
    return np.sign(anomaly) * (np.abs(anomaly) > threshold)


def phase_counts(index, window, phases=PHASES):
    """Count phase occurrences in a centred rolling window along time.

    Args:
        index (xr.DataArray): Phase index of -1, 0, 1.
        window (int): Width of the centred rolling window.
        phases (xr.DataArray): Phase values to count, on a ``phase`` dimension.

    Returns:
        xr.DataArray: Counts with a new ``phase`` dimension; NaN where the window is incomplete.
    """
    count_da = (index == phases).rolling(time=window, center=True).sum()
    count_da.name = 'count'
    return count_da.isel(time=slice(window // 2, -window // 2))
```

```python
import numpy as np
from scipy.stats import gaussian_kde


def kde_axis(vmin: float, vmax: float, n: int) -> np.ndarray:
    """Evaluation points for a KDE.

    vmin, vmax : float
        Bounds of the axis, in data units.
    n : int
        Number of points.

    Returns
    -------
    np.ndarray, shape (n,)
    """
    return np.linspace(vmin, vmax, n)


def kde(samples: np.ndarray, x: np.ndarray, bw_method=None) -> np.ndarray:
    """Gaussian KDE of `samples` evaluated at `x`.

    samples : np.ndarray, shape (m,)
        1-D sample array.
    x : np.ndarray, shape (n,)
        Evaluation points, e.g. from `kde_axis`.
    bw_method : str | float | callable | None
        Passed straight to scipy.stats.gaussian_kde.

    Returns
    -------
    np.ndarray, shape (n,)
        Density at each point of `x`.
    """
    return gaussian_kde(samples, bw_method=bw_method)(x)
```

```python
def plot_kde_panel(ax, dens, era5_kde, x):
    """Plot ensemble member KDEs and the ERA5 KDE on one axis.

    Args:
        ax (matplotlib.axes.Axes): Axis to draw on.
        dens (xr.DataArray): Ensemble densities with dims (member, x).
        era5_kde (np.ndarray): ERA5 density evaluated at x.
        x (np.ndarray): Evaluation grid.

    Returns:
        tuple[Line2D, Line2D]: Handles for one member line and the ERA5 line.
    """
    members = dens.plot(ax=ax, hue="member", x="x", add_legend=False, color="grey", alpha=0.6)
    (era5,) = ax.plot(x, era5_kde, color="blue", lw=2)
    return members[0], era5
```

# Percent mean from extremes

Rank decomposition of the forced mean shift, season by season.
Sorting does not change a mean, so for two samples of equal length
mean(hist) - mean(histnat) equals the mean over i of x_(i) - y_(i), where
x_(i) and y_(i) are the i-th smallest values of each sample. The mean shift
therefore partitions exactly across ranks, which is what makes quantile_diff
an exact decomposition rather than a suggestive picture, and what lets it say
where in the distribution the warming lives.
Season labelling follows the convention that a season is named for the year
its January falls in, so DJF 2003/04 has season_year 2004.

## Vectorised

```python


# ------------------------------------------------------------------- plotting

SEASON_COLOURS = {
    'DJF': 'firebrick',
    'MAM': 'darkorange',
    'JJA': 'steelblue',
    'SON': 'seagreen',
}

BAND_COLOURS = {
    'diff_cold_tail': 'steelblue',
    'diff_middle': 'dimgrey',
    'diff_hot_tail': 'firebrick',
}

TRIM_COLOURS = {
    'mean_diff_trim_hot': {'colour': 'firebrick', 'label': 'hottest removed'},
    'mean_diff_trim_cold': {'colour': 'steelblue', 'label': 'coldest removed'},
}

N_CONNECTORS = 12  # rank matched pairs drawn on the method figure
ZOOM_WINDOW = (0.45, 0.55)  # quantile range of the method figure inset
EDGE_TRIM = 0.005  # quantiles hidden at each edge, where order statistics dominate

RECORD_AXES = {
    'quantile_diff': {
        'x': 'quantile_level',
        'x_label': 'quantile',
        'colourbar_label': 'hist - histnat (K)',
    },
    'mean_diff_trim_hot': {
        'x': 'n_removed',
        'x_label': 'hottest days removed',
        'colourbar_label': 'remaining mean shift (K)',
    },
    'mean_diff_trim_cold': {
        'x': 'n_removed',
        'x_label': 'coldest days removed',
        'colourbar_label': 'remaining mean shift (K)',
    },
}


def tidy(ax):
    """
    Drop the top and right spines so the data is not boxed in.
    Parameters
    ----------
    ax (matplotlib.axes.Axes) :Axes to tidy.
    Returns
    -------
    ax (matplotlib.axes.Axes) :The same axes.
    """
    ax.spines[['top', 'right']].set_visible(False)
    return ax


def plot_method(averaged, season='DJF', axes=None):
    """
    Three panels walking through the method on one season. The first shows the
    two quantile functions with a handful of rank matched pairs joined, which
    is the step that makes two unpaired ensembles comparable: the i-th coldest
    day of one is set against the i-th coldest of the other, not against the
    same calendar day. The second shows those differences, which by the sorting
    identity average to exactly the difference in means, so the panel is the
    mean shift taken apart by where in the distribution it comes from. The
    third shows the mean shift remaining as days are deleted from each end,
    which is the original framing of the question and the cumulative integral
    of the second panel.
    Parameters
    ----------
    averaged (xr.Dataset) :Decomposition averaged over season_year, carrying the quantile functions and both trimmed mean curves.
    season (str) :Season to illustrate.
    axes (array of matplotlib.axes.Axes, optional) :Three axes to draw into, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    one = averaged.sel(season=season)
    if axes is None:
        _, axes = plt.subplots(1, 3, figsize=(14, 4.2), layout='constrained')

    quantile = one['quantile_level']
    n_rank = one.sizes['rank']
    connectors = np.linspace(0, n_rank - 1, N_CONNECTORS).astype(int)
    for rank in connectors:
        axes[0].plot(
            [quantile[rank]] * 2,
            [one.histnat_q[rank], one.hist_q[rank]],
            color='0.6', lw=0.8, zorder=1,
        )
    axes[0].plot(quantile, one.histnat_q, color='steelblue', lw=1.6, label='histnat', zorder=2)
    axes[0].plot(quantile, one.hist_q, color='firebrick', lw=1.6, label='hist', zorder=2)
    axes[0].set_xlabel('quantile')
    axes[0].set_ylabel('temperature (degC)')
    axes[0].set_title('1. match rank for rank', loc='left')
    axes[0].legend(frameon=False, loc='upper left')

    # The shift is around a kelvin against a range of tens of kelvin, so the
    # two curves are indistinguishable at full scale and the inset carries the
    # actual point of the panel.
    window = slice(int(ZOOM_WINDOW[0] * n_rank), int(ZOOM_WINDOW[1] * n_rank))
    inset = axes[0].inset_axes([0.52, 0.1, 0.44, 0.36])
    for rank in np.linspace(window.start, window.stop - 1, 5).astype(int):
        inset.plot(
            [quantile[rank]] * 2,
            [one.histnat_q[rank], one.hist_q[rank]],
            color='0.6', lw=1.0, zorder=1,
        )
    inset.plot(quantile[window], one.histnat_q[window], color='steelblue', lw=1.4, zorder=2)
    inset.plot(quantile[window], one.hist_q[window], color='firebrick', lw=1.4, zorder=2)
    inset.set_xticks(ZOOM_WINDOW)
    inset.tick_params(labelsize=8)
    axes[0].indicate_inset_zoom(inset, edgecolor='0.4')

    axes[1].fill_between(quantile, 0, one.quantile_diff, color='firebrick', alpha=0.18)
    axes[1].plot(quantile, one.quantile_diff, color='firebrick', lw=1.2)
    axes[1].axhline(float(one.mean_diff), color='k', lw=0.9, ls='--')
    axes[1].annotate(
        f'mean shift {float(one.mean_diff):.2f} K',
        xy=(0.03, float(one.mean_diff)), xytext=(0, 6),
        textcoords='offset points', fontsize=9,
    )
    axes[1].set_xlabel('quantile')
    axes[1].set_ylabel('hist - histnat (K)')
    axes[1].set_title('2. difference at each rank', loc='left')

    for name, style in TRIM_COLOURS.items():
        axes[2].plot(
            one.n_removed, one[name], color=style['colour'], lw=1.2,
            label=style['label'],
        )
    axes[2].axhline(float(one.mean_diff), color='k', lw=0.9, ls='--')
    axes[2].axhline(0, color='k', lw=0.7)
    axes[2].set_xlabel('days removed')
    axes[2].set_ylabel('remaining mean shift (K)')
    axes[2].set_title('3. trim from each end', loc='left')
    axes[2].legend(frameon=False)

    for ax in axes:
        tidy(ax)
    return axes


def plot_results(averaged, decomposition, axes=None):
    """
    Four panels carrying the result. The top row is distribution shape averaged
    over recent seasons, since a single season is far too noisy to read shape
    from, and the bottom row is the full record. Warming against quantile
    flat on its dashed reference means the distribution translated, the share
    curve on the diagonal means extremes contributed exactly their proportion
    of the mean change and no more, and the standard deviation ratio flat at
    one means the spread never changed. Those three say the same thing three
    independent ways, which is what makes the conclusion defensible.
    Parameters
    ----------
    averaged (xr.Dataset) :Decomposition averaged over season_year, carrying quantile_diff, mean_diff and share.
    decomposition (xr.Dataset) :Full record output of rank_decomposition_full_set, carrying mean_diff and qq_slope.
    axes (array of matplotlib.axes.Axes, optional) :Four axes to draw into, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    if axes is None:
        _, axes = plt.subplots(2, 2, figsize=(12, 8.5), layout='constrained')
    axes = np.asarray(axes).ravel()

    for season, colour in SEASON_COLOURS.items():
        shape = averaged.sel(season=season)

        # The outermost ranks are order statistics whose sampling error is many
        # times the signal, and left in they set the y range for the panel.
        level = shape['quantile_level']
        inner = (level > EDGE_TRIM) & (level < 1 - EDGE_TRIM)
        axes[0].plot(
            level.where(inner), shape.quantile_diff.where(inner),
            color=colour, lw=1.1, label=season,
        )
        axes[0].axhline(float(shape.mean_diff), color=colour, lw=0.7, ls='--')
        axes[1].plot(shape.hottest_fraction, shape.share, color=colour, lw=1.4, label=season)
        axes[2].plot(decomposition.season_year, decomposition.mean_diff.sel(season=season), color=colour, lw=0.9, label=season)
        axes[3].plot(decomposition.season_year, decomposition.qq_slope.sel(season=season), color=colour, lw=0.9, label=season)

    axes[0].set_xlabel('quantile')
    axes[0].set_ylabel('hist - histnat (K)')
    axes[0].set_title('warming by position in the distribution', loc='left')
    axes[0].legend(frameon=False, ncol=4, fontsize=9)

    axes[1].plot([0, 1], [0, 1], color='k', lw=0.9, ls='--')
    axes[1].annotate(
        'equal share', xy=(0.62, 0.62), xytext=(6, -14),
        textcoords='offset points', rotation=45, fontsize=9, color='0.3',
    )
    axes[1].set_xlabel('hottest fraction of days')
    axes[1].set_ylabel('share of mean shift')
    axes[1].set_title('share carried by the hottest days', loc='left')

    axes[2].axhline(0, color='k', lw=0.7)
    axes[2].set_xlabel('season year')
    axes[2].set_ylabel('mean shift (K)')
    axes[2].set_title('when the shift emerges', loc='left')

    axes[3].axhline(1, color='k', lw=0.9, ls='--')
    axes[3].set_xlabel('season year')
    axes[3].set_ylabel('hist / histnat standard deviation')
    axes[3].set_title('spread ratio through the record', loc='left')

    for ax in axes:
        tidy(ax)
    return axes


def plot_bands(decomposition, axes=None):
    """
    Cold tail, middle and hot tail mean differences through the record, one
    panel per season. Under a uniform shift the three track each other, and a
    band pulling systematically away from the others is where the distribution
    is changing shape. The cold tail is far noisier than the middle because it
    averages a tenth of the ranks in the sparsest part of the distribution, so
    its larger swings are not evidence of anything on their own.
    Parameters
    ----------
    decomposition (xr.Dataset) :Decomposition carrying the band mean differences and a season dimension.
    axes (array of matplotlib.axes.Axes, optional) :One axes per season, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    if axes is None:
        _, axes = plt.subplots(
            len(SEASON_COLOURS), 1, figsize=(8, 10), sharex=True, sharey=True,
        )

    for ax, season in zip(axes, SEASON_COLOURS):
        one = decomposition.sel(season=season)
        for name, colour in BAND_COLOURS.items():
            ax.plot(
                one.season_year, one[name], color=colour, lw=0.9,
                label=name.replace('diff_', '').replace('_', ' '),
            )
        ax.axhline(0, color='k', lw=0.7)
        ax.set_ylabel(f'{season} (K)')
        tidy(ax)

    axes[0].legend(frameon=False, ncol=3, fontsize=9)
    axes[-1].set_xlabel('season year')
    return axes


def plot_record(decomposition, name='quantile_diff', season='DJF', ax=None):
    """
    One diagnostic through the whole record as a heatmap, season year against
    the diagnostic's own x axis with the value in colour. For quantile_diff a
    row is one season's warming by position in the distribution, so horizontal
    banding that deepens towards the present is a uniform shift emerging, and a
    persistent vertical stripe is one part of the distribution behaving
    differently from the rest. For the trimmed mean curves a row is the shift
    remaining as days are deleted from one end, so a row fading to white is a
    season whose shift trimming can remove. robust clips the colour scale at
    the second and ninety eighth percentiles so the emergent decades do not
    saturate it, and the scale is centred on zero so sign is always legible.
    Parameters
    ----------
    decomposition (xr.Dataset) :Output of rank_decomposition_full_set.
    name (str) :Variable to draw, one of the keys of RECORD_AXES.
    season (str) :Season to draw.
    ax (matplotlib.axes.Axes, optional) :Axes to draw into, created if not given.
    Returns
    -------
    ax (matplotlib.axes.Axes) :The axes drawn into.
    """
    settings = RECORD_AXES[name]
    one = decomposition[name].sel(season=season)
    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))

    one.plot(
        x=settings['x'], y='season_year', center=0, robust=True, ax=ax,
        cbar_kwargs={'label': settings['colourbar_label']},
    )
    ax.set_xlabel(settings['x_label'])
    ax.set_ylabel('season year')
    ax.set_title(season, loc='left')
    tidy(ax)
    return ax
```

```python
"""
Rank decomposition of the forced mean shift, season by season.
Sorting does not change a mean, so for two samples of equal length
mean(hist) - mean(histnat) equals the mean over i of x_(i) - y_(i), where
x_(i) and y_(i) are the i-th smallest values of each sample. The mean shift
therefore partitions exactly across ranks, which is what makes quantile_diff
an exact decomposition rather than a suggestive picture, and what lets it say
where in the distribution the warming lives.
Season labelling follows the convention that a season is named for the year
its January falls in, so DJF 2003/04 has season_year 2004. Only DJF crosses a
year boundary, so only DJF takes that offset.
"""

import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

SEASONS = ['DJF', 'MAM', 'JJA', 'SON']
DAYS_PER_SEASON = 90  # 360 day calendar, so every season is exactly 90 steps
MIN_RETAINED = 30  # trimmed means over fewer ranks than this are unusable
N_QUANTILE_LEVELS = 1000  # grid size when not matching rank for rank

# Quantile ranges reduced to a single mean difference each. These are the only
# form of the decomposition that survives to a map, since 4050 ranks per
# gridpoint cannot be stored continent wide.
BANDS = {
    'cold_tail': {'lower': 0.0, 'upper': 0.1},
    'middle': {'lower': 0.25, 'upper': 0.75},
    'hot_tail': {'lower': 0.9, 'upper': 1.0},
}


# ---------------------------------------------------------------- preparation

def add_season_year(da, season):
    """
    Label each timestep with the year its season ends in. Only DJF crosses a
    year boundary, so only there does December need pushing forward to sit with
    the following January and February. Filtering on time.dt.year directly
    would otherwise mix the tail of one summer with the head of the next. For
    the other three seasons the label is just the calendar year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data with a cftime time coordinate.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    labelled (xr.DataArray) :Input with a season_year coordinate added along time.
    """
    rolls_over = (da.time.dt.month == 12) & (season == 'DJF')
    labelled = da.assign_coords(season_year=da.time.dt.year + rolls_over)
    return labelled


def full_season_years(da):
    """
    Find season years holding a complete season, used to drop the part seasons
    at each end of the record where the opening or closing months fall outside
    the data.
    Parameters
    ----------
    da (xr.DataArray) :Data carrying a season_year coordinate along time.
    Returns
    -------
    complete (np.ndarray) :Season years containing exactly DAYS_PER_SEASON timesteps.
    """
    counts = da.season_year.groupby('season_year').count()
    complete = counts.season_year.values[counts.values == DAYS_PER_SEASON]
    return complete


def match_dims(da1, da2, season):
    """
    Reduce both records to one season over a common set of members and complete
    seasons. Rank matching is meaningless unless the two samples are the same
    size, so equal sample size is enforced here rather than assumed downstream,
    where a mismatch would silently compare offset ranks.
    Parameters
    ----------
    da1 (xr.DataArray) :Daily temperature with member and time, conventionally the forced record.
    da2 (xr.DataArray) :Daily temperature to compare against, on the same dimensions.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    da1, da2 (xr.DataArray) :The two records subset to matching members and complete seasons, loaded into memory.
    """
    da1 = add_season_year(
        da1.sel(time=da1.time.dt.season == season), season
    ).sortby('time')
    da2 = add_season_year(
        da2.sel(time=da2.time.dt.season == season), season
    ).sortby('time')

    # The two sets of members share no initial conditions, so this buys equal
    # sample size for rank matching and nothing more. It is not physical pairing.
    members = np.intersect1d(da1.member, da2.member)
    da1 = da1.sel(member=members)
    da2 = da2.sel(member=members)

    years = np.intersect1d(full_season_years(da1), full_season_years(da2))
    da1 = da1.sel(time=da1.season_year.isin(years)).load()
    da2 = da2.sel(time=da2.season_year.isin(years)).load()

    # to_year_day assigns day numbers by tiling, which is only valid if the
    # sorted record really is contiguous blocks of DAYS_PER_SEASON.
    expected = np.repeat(years, DAYS_PER_SEASON)
    assert (da1.season_year.values == expected).all()
    assert (da2.season_year.values == expected).all()
    return da1, da2


def match_dims_by_season(da1, da2, seasons=SEASONS):
    """
    Run match_dims for each season and stack the results on a season dimension,
    giving one array per record with season, season_year, member and day. The
    decomposition never names season, so once the records are shaped this way
    every season is decomposed independently with no further looping.
    Parameters
    ----------
    da1 (xr.DataArray) :Daily temperature with member and time, conventionally the forced record.
    da2 (xr.DataArray) :Daily temperature to compare against, on the same dimensions.
    seasons (list of str) :Season names to include, defaulting to all four.
    Returns
    -------
    da1_prepared, da2_prepared (xr.DataArray) :The two records with dimensions season, season_year, member and day.
    """
    da1_by_season = []
    da2_by_season = []
    for season in seasons:
        da1_season, da2_season = match_dims(da1, da2, season)
        da1_by_season.append(to_year_day(da1_season))
        da2_by_season.append(to_year_day(da2_season))

    # join='inner' keeps only the season years every season has, since DJF
    # loses one at each end. Outer joining would pad the others with NaN, and
    # NaN sorts to the hot end of the distribution and would corrupt the
    # ranking rather than just leaving a gap.
    season_dim = xr.DataArray(seasons, dims='season', name='season')
    da1_prepared = xr.concat(da1_by_season, dim=season_dim, join='inner')
    da2_prepared = xr.concat(da2_by_season, dim=season_dim, join='inner')
    return da1_prepared, da2_prepared


# ------------------------------------------------------------------- ranking

def to_year_day(da):
    """
    Reshape the flat time axis into separate season_year and day dimensions.
    Purely structural with no arithmetic, but it is what allows each season to
    be sorted independently in one vectorised call rather than through a
    groupby, which would build a separate dask graph per year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data carrying a season_year coordinate, as returned by match_dims.
    Returns
    -------
    reshaped (xr.DataArray) :Input with time replaced by season_year and day.
    """
    n_years = np.unique(da.season_year.values).size
    reshaped = (
        da.assign_coords(
            day=('time', np.tile(np.arange(DAYS_PER_SEASON), n_years))
        )
        .set_index(time=('season_year', 'day'))
        .unstack('time')
    )
    return reshaped


def sort_pooled(arr):
    """
    Pool the trailing two axes into one and sort it ascending. Operating on the
    last two axes is what lets apply_ufunc broadcast this over any leading
    dimensions, so the same kernel works at one gridpoint or across a map.
    Parameters
    ----------
    arr (np.ndarray) :Array whose final two axes are member and day.
    Returns
    -------
    ordered (np.ndarray) :Same leading axes, with the trailing two replaced by one sorted axis.
    """
    ordered = np.sort(arr.reshape(arr.shape[:-2] + (-1,)), axis=-1)
    return ordered


def rank_sort(da):
    """
    Build the empirical quantile function by pooling every member and every day
    into a single sample and sorting it. Members are treated as alternative
    realisations of the same season, so the resulting distribution describes
    the days that could have occurred in that year rather than the days one
    member happened to produce. Any dimensions other than member and day, such
    as season_year or season, are left alone and sorted independently.
    Parameters
    ----------
    da (xr.DataArray) :Data with member and day dimensions, as returned by to_year_day.
    Returns
    -------
    ranked (xr.DataArray) :Leading dimensions of da with member and day replaced by rank, ascending along rank, carrying a quantile_level coordinate giving the plotting position of each rank.
    """
    n_sample = da.sizes['member'] * da.sizes['day']

    # Naming member and day as core dims moves them to the trailing axes, which
    # is exactly what sort_pooled expects to flatten.
    ranked = xr.apply_ufunc(
        sort_pooled,
        da,
        input_core_dims=[['member', 'day']],
        output_core_dims=[['rank']],
    )

    # quantile_level, not quantile, because attribute access resolves methods
    # before coordinates and DataArray.quantile would shadow it.
    ranked = ranked.assign_coords(
        quantile_level=('rank', (np.arange(n_sample) + 0.5) / n_sample)
    )
    return ranked


def quantile_grid_function(da, pool_dims):
    """
    Empirical quantile function evaluated on a fixed grid of probability levels
    rather than by sorting. Sorting requires the two samples to be the same
    size, since it matches the i-th smallest of one against the i-th smallest
    of the other, and that constraint is what forces equal member counts and
    equal season lengths. Evaluating both sides on a shared probability grid
    lifts it, which is what allows records of unequal length, calendars where a
    season is sometimes 91 days, and a reference pooled over far more values
    than the sample it is compared against. The cost is that the decomposition
    becomes a very close approximation rather than an exact identity, because
    mean_diff is then a mean over grid points instead of over every value.
    Parameters
    ----------
    da (xr.DataArray) :Data with member and day dimensions, as returned by to_year_day.
    pool_dims (list of str) :Dimensions pooled before evaluating quantiles, member and day for a per season distribution, or member, day and season_year for a reference pooled over the whole record.
    Returns
    -------
    evaluated (xr.DataArray) :Dimensions of da less pool_dims, plus rank indexing the probability grid and carrying a quantile_level coordinate.
    """
    levels = (np.arange(N_QUANTILE_LEVELS) + 0.5) / N_QUANTILE_LEVELS

    # quantile names its new dimension after itself, so it is renamed to rank
    # to keep every downstream function unchanged. In this path rank indexes
    # probability levels rather than sample order.
    evaluated = (
        da.quantile(levels, dim=pool_dims)
        .rename(quantile='rank')
        .drop_vars('rank')
        .assign_coords(quantile_level=('rank', levels))
    )
    return evaluated


# ------------------------------------------- summaries of the decomposition

def trimmed_mean_diff(quantile_diff, from_hot_end):
    """
    Mean rank difference remaining after deleting a growing number of days from
    one end of the distribution. This is the cumulative form of the original
    question, which asked how many extreme events would have to be removed
    before the mean returns to the counterfactual. Both directions are offered
    because whichever tail is least changed is the wrong one to delete.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, with a rank dimension.
    from_hot_end (bool) :True deletes the hottest days first, False the coldest.
    Returns
    -------
    curve (xr.DataArray) :Leading dimensions of quantile_diff with rank replaced by n_removed, which counts deleted days and runs upwards from zero.
    """
    diff = quantile_diff.transpose(..., 'rank')

    # Ranks ascend, so a cumulative sum accumulates from the cold end and
    # therefore describes deleting the hottest days. Reversing the input first
    # flips which end is being deleted.
    values = diff.values if from_hot_end else diff.values[..., ::-1]

    n_rank = diff.sizes['rank']
    retained = np.arange(1, n_rank + 1)

    # cumsum is indexed by days kept, so both it and the divisor are reversed
    # to give an output indexed by days removed, ascending from zero.
    means = np.cumsum(values, axis=-1)[..., ::-1] / retained[::-1]

    curve = xr.DataArray(
        means,
        dims=diff.dims[:-1] + ('n_removed',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(n_removed=('n_removed', np.arange(n_rank)))

    # Past this point the trimmed mean is taken over a handful of ranks and its
    # sampling error swamps the signal, so the tail is masked rather than
    # plotted as though it meant something.
    curve = curve.where(curve.n_removed <= n_rank - MIN_RETAINED)
    return curve


def share_of_mean_shift(quantile_diff):
    """
    Fraction of the total mean shift carried by the hottest fraction of the
    distribution, which is the statistic the whole diagnostic was built to
    answer. Under a uniform shift the hottest tenth of days carries a tenth of
    the mean change, so the curve lies on the diagonal, and extremes carrying
    more than their share bows it above. This asks about proportion rather than
    about a threshold being reached, so unlike a crossing point it has an
    answer for every season rather than only for those where trimming happens
    to cancel the shift. It divides by the total shift, so it is meaningless in
    seasons where that total is near zero and the forced signal has not
    emerged. Rank differences can be negative, so unlike the precipitation
    version of this diagnostic the share is not confined between zero and one,
    and a value above one means part of the distribution cooled.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, with a rank dimension.
    Returns
    -------
    share (xr.DataArray) :Leading dimensions of quantile_diff with rank replaced by hottest_fraction, running from one rank up to the whole distribution.
    """
    diff = quantile_diff.transpose(..., 'rank')
    n_rank = diff.sizes['rank']

    # Ranks ascend, so the cumulative sum is reversed to accumulate from the
    # hot end. The final element is the total shift, which normalises the rest.
    cumulative = np.cumsum(diff.values[..., ::-1], axis=-1)
    share = cumulative / cumulative[..., -1:]

    share = xr.DataArray(
        share,
        dims=diff.dims[:-1] + ('hottest_fraction',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(
        hottest_fraction=('hottest_fraction', np.arange(1, n_rank + 1) / n_rank)
    )
    return share


def quantile_quantile_fit(hist_q, histnat_q):
    """
    Least squares fit of the hist quantile function against the histnat
    quantile function, which separates a change in location from a change in
    spread. Writing x for histnat_q and y for hist_q along rank, the slope is
    sum((x_i - xbar)(y_i - ybar)) / sum((x_i - xbar)^2) and the intercept is
    ybar - slope * xbar, which is a degree one polyfit expressed as covariance
    over variance so that it vectorises over the leading dimensions without a
    Python loop. Because both inputs are sorted this is the quantile quantile
    fit, so the slope is effectively the ratio of standard deviations: one
    means the distribution simply translated and every quantile warmed by the
    same amount, above one means it stretched so extremes warmed more, below
    one means it compressed.
    The intercept is the fitted value at zero degrees, which for polar data is
    tens of degrees outside the sample, so small slope errors are levered into
    large intercept errors. Use mean_diff for the location shift, not this.
    Parameters
    ----------
    hist_q (xr.DataArray) :Sorted quantile function of the historical experiment, with a rank dimension, as returned by rank_sort.
    histnat_q (xr.DataArray) :Sorted quantile function of the natural forcing only experiment.
    Returns
    -------
    slope, intercept (xr.DataArray) :Fit coefficients over the leading dimensions of the inputs.
    """
    histnat_q_centred = histnat_q - histnat_q.mean('rank')
    slope = (
        histnat_q_centred * (hist_q - hist_q.mean('rank'))
    ).sum('rank') / (histnat_q_centred ** 2).sum('rank')
    intercept = hist_q.mean('rank') - slope * histnat_q.mean('rank')
    return slope, intercept


def band_mean_diffs(quantile_diff):
    """
    Mean rank difference within each quantile range in BANDS, reducing the
    decomposition from thousands of ranks to one number per band. These are the
    only summaries compact enough to carry to a map, and comparing them across
    seasons is how a uniform shift shows itself, since under translation the
    tails and the middle track each other. The bands cover different numbers of
    ranks in regions of different density, so their sampling errors differ by
    several times and the curves are not comparable at face value.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, carrying a quantile_level coordinate along rank.
    Returns
    -------
    diffs (dict of str to xr.DataArray) :Band name prefixed with diff_, mapped to the mean difference over that band with the rank dimension reduced away.
    """
    level = quantile_diff['quantile_level']

    # where masks to the band and mean skips the resulting NaN, so each band is
    # averaged over its own ranks only.
    diffs = {}
    for name, band in BANDS.items():
        within = (level >= band['lower']) & (level <= band['upper'])
        diffs[f'diff_{name}'] = quantile_diff.where(within).mean('rank')
    return diffs


# --------------------------------------------------------------- decomposition

def rank_decomposition(da1, da2, use_quantile_grid=False,
                       pooled_reference=False):
    """
    Decompose the hist minus histnat mean shift across the distribution. This
    returns only the decomposition itself, since quantile_diff and mean_diff
    are exact consequences of the sorting identity whereas every summary of
    them is a choice. The trimmed mean curves, the share, the
    quantile quantile fit and the band means are separate functions to be added
    explicitly, so that what a given results object contains is visible at the
    call site rather than buried here. Only member and day are consumed, so any
    other dimension present, including season, passes through untouched and
    each of its elements is decomposed independently.
    Parameters
    ----------
    da1 (xr.DataArray) :Prepared data with member and day dimensions, conventionally the forced record.
    da2 (xr.DataArray) :Prepared data to compare against, on the same dimensions.
    use_quantile_grid (bool) :Evaluate both distributions on a shared probability grid instead of matching rank for rank, which allows the two samples to differ in size at the cost of the decomposition becoming approximate.
    pooled_reference (bool) :Compare against histnat pooled over its whole record rather than against the matching season year, which removes the sampling noise contributed by the reference. Requires use_quantile_grid since the pooled sample is far larger.
    Returns
    -------
    results (xr.Dataset) :The two quantile functions, their rank difference and the mean shift it sums to.
    """
    assert use_quantile_grid or not pooled_reference, (
        'a pooled reference holds many more values than one hist season, so '
        'rank matching cannot align them. Set use_quantile_grid=True.'
    )

    # Pooling season_year into the reference collapses it, so histnat_q loses
    # its season dimension and broadcasts against every hist season.
    reference_dims = ['member', 'day']
    if pooled_reference:
        reference_dims = reference_dims + ['season_year']

    if use_quantile_grid:
        hist_q = quantile_grid_function(da1, ['member', 'day'])
        histnat_q = quantile_grid_function(da2, reference_dims)
    else:
        hist_q = rank_sort(da1)
        histnat_q = rank_sort(da2)

    # The i-th coldest hist day minus the i-th coldest histnat day. Not the
    # same days, the same positions in each distribution.
    quantile_diff = hist_q - histnat_q

    # Exactly mean(hist) - mean(histnat) by the sorting identity, so this is
    # the quantity trimming is trying to eliminate.
    mean_diff = quantile_diff.mean('rank')

    results = xr.Dataset({
        # The quantile functions are kept rather than discarded because a 1 K
        # difference means something different at -45 degC than at -10 degC,
        # and only the absolute curves say which.
        'hist_q': hist_q,
        'histnat_q': histnat_q,
        'quantile_diff': quantile_diff,
        'mean_diff': mean_diff,
    })
    return results


def rank_decomposition_full_set(da1, da2, use_quantile_grid=False,
                                pooled_reference=False):
    """
    Run the decomposition and add every summary of it that the figures use, so
    that one call produces a complete results object. rank_decomposition stays
    separate and returns only the exact decomposition, since the summaries
    below are choices rather than consequences, and dropping one should be a
    matter of deleting a line here rather than editing the core.
    Parameters
    ----------
    da1 (xr.DataArray) :Prepared data with member and day dimensions, conventionally the forced record.
    da2 (xr.DataArray) :Prepared data to compare against, on the same dimensions.
    use_quantile_grid (bool) :Passed to rank_decomposition.
    pooled_reference (bool) :Passed to rank_decomposition.
    Returns
    -------
    full_set (xr.Dataset) :The decomposition together with both trimmed mean curves, the share of the mean shift, the standard deviation ratio and the band mean differences.
    """
    results = rank_decomposition(
        da1,
        da2,
        use_quantile_grid=use_quantile_grid,
        pooled_reference=pooled_reference,
    )
    quantile_diff = results.quantile_diff

    mean_diff_trim_hot = trimmed_mean_diff(quantile_diff, from_hot_end=True)
    mean_diff_trim_cold = trimmed_mean_diff(quantile_diff, from_hot_end=False)
    share = share_of_mean_shift(quantile_diff)
    band_diffs = band_mean_diffs(quantile_diff)

    # The intercept is the fit extrapolated to zero degrees, tens of degrees
    # outside polar data, so it is discarded and mean_diff carries the shift.
    qq_slope, _ = quantile_quantile_fit(results.hist_q, results.histnat_q)

    full_set = results.assign(
        mean_diff_trim_hot=mean_diff_trim_hot,
        mean_diff_trim_cold=mean_diff_trim_cold,
        share=share,
        qq_slope=qq_slope,
        **band_diffs,
    )
    return full_set


```

```python


# ------------------------------------------------------------ distribution fit

def quantile_quantile_fit(hist_q, histnat_q):
    """
    Least squares fit of the hist quantile function against the histnat
    quantile function, which separates a change in location from a change in
    spread. Writing x for histnat_q and y for hist_q along rank, the slope is
    sum((x_i - xbar)(y_i - ybar)) / sum((x_i - xbar)^2) and the intercept is
    ybar - slope * xbar, which is a degree one polyfit expressed as covariance
    over variance so that it vectorises over the leading dimensions without a
    Python loop. Because both inputs are sorted this is the quantile quantile
    fit, so the slope is effectively the ratio of standard deviations: one
    means the distribution simply translated and every quantile warmed by the
    same amount, above one means it stretched so extremes warmed more, below
    one means it compressed.
    The intercept is the fitted value at zero degrees, which for polar data is
    tens of degrees outside the sample, so small slope errors are levered into
    large intercept errors. Use mean_diff for the location shift, not this.
    Parameters
    ----------
    hist_q (xr.DataArray) :Sorted quantile function of the historical experiment, with a rank dimension, as returned by rank_sort.
    histnat_q (xr.DataArray) :Sorted quantile function of the natural forcing only experiment.
    Returns
    -------
    slope, intercept (xr.DataArray) :Fit coefficients over the leading dimensions of the inputs.
    """
    histnat_q_centred = histnat_q - histnat_q.mean('rank')
    slope = (
        histnat_q_centred * (hist_q - hist_q.mean('rank'))
    ).sum('rank') / (histnat_q_centred ** 2).sum('rank')
    intercept = hist_q.mean('rank') - slope * histnat_q.mean('rank')
    return slope, intercept


def band_mean_diffs(quantile_diff):
    """
    Mean rank difference within each quantile range in BANDS, reducing the
    decomposition from thousands of ranks to one number per band. These are the
    only summaries compact enough to carry to a map, and comparing them across
    seasons is how a uniform shift shows itself, since under translation the
    tails and the middle track each other. The bands cover different numbers of
    ranks in regions of different density, so their sampling errors differ by
    several times and the curves are not comparable at face value.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, carrying a quantile_level coordinate along rank.
    Returns
    -------
    diffs (dict of str to xr.DataArray) :Band name prefixed with diff_, mapped to the mean difference over that band with the rank dimension reduced away.
    """
    level = quantile_diff['quantile_level']

    # where masks to the band and mean skips the resulting NaN, so each band is
    # averaged over its own ranks only.
    diffs = {}
    for name, band in BANDS.items():
        within = (level >= band['lower']) & (level <= band['upper'])
        diffs[f'diff_{name}'] = quantile_diff.where(within).mean('rank')
    return diffs


# --------------------------------------------------------------- decomposition

def rank_decomposition(hist, histnat, use_quantile_grid=False,
                       pooled_reference=False):
    """
    Decompose the hist minus histnat mean shift across the distribution. This
    returns only the decomposition itself, since quantile_diff and mean_diff
    are exact consequences of the sorting identity whereas every summary of
    them is a choice. The trimmed mean curves, the crossing points, the
    quantile quantile fit and the band means are separate functions to be added
    explicitly, so that what a given results object contains is visible at the
    call site rather than buried here. Only member and day are consumed, so any
    other dimension present, including season, passes through untouched and
    each of its elements is decomposed independently.
    Parameters
    ----------
    hist (xr.DataArray) :Prepared historical data with member and day dimensions, as returned by to_year_day.
    histnat (xr.DataArray) :Prepared natural forcing only data on the same dimensions.
    use_quantile_grid (bool) :Evaluate both distributions on a shared probability grid instead of matching rank for rank, which allows the two samples to differ in size at the cost of the decomposition becoming approximate.
    pooled_reference (bool) :Compare against histnat pooled over its whole record rather than against the matching season year, which removes the sampling noise contributed by the reference. Requires use_quantile_grid since the pooled sample is far larger.
    Returns
    -------
    results (xr.Dataset) :The two quantile functions, their rank difference and the mean shift it sums to.
    """
    assert use_quantile_grid or not pooled_reference, (
        'a pooled reference holds many more values than one hist season, so '
        'rank matching cannot align them. Set use_quantile_grid=True.'
    )

    # Pooling season_year into the reference collapses it, so histnat_q loses
    # its season dimension and broadcasts against every hist season.
    reference_dims = ['member', 'day']
    if pooled_reference:
        reference_dims = reference_dims + ['season_year']

    if use_quantile_grid:
        hist_q = quantile_grid_function(hist, ['member', 'day'])
        histnat_q = quantile_grid_function(histnat, reference_dims)
    else:
        hist_q = rank_sort(hist)
        histnat_q = rank_sort(histnat)

    # The i-th coldest hist day minus the i-th coldest histnat day. Not the
    # same days, the same positions in each distribution.
    quantile_diff = hist_q - histnat_q

    # Exactly mean(hist) - mean(histnat) by the sorting identity, so this is
    # the quantity trimming is trying to eliminate.
    mean_diff = quantile_diff.mean('rank')

    results = xr.Dataset({
        # The quantile functions are kept rather than discarded because a 1 K
        # difference means something different at -45 degC than at -10 degC,
        # and only the absolute curves say which.
        'hist_q': hist_q,
        'histnat_q': histnat_q,
        'quantile_diff': quantile_diff,
        'mean_diff': mean_diff,
    })
    return results


# ------------------------------------------------------------------- plotting

SEASON_COLOURS = {
    'DJF': 'firebrick',
    'MAM': 'darkorange',
    'JJA': 'steelblue',
    'SON': 'seagreen',
}

BAND_COLOURS = {
    'diff_cold_tail': 'steelblue',
    'diff_middle': 'dimgrey',
    'diff_hot_tail': 'firebrick',
}

TRIM_COLOURS = {
    'mean_diff_trim_hot': {'colour': 'firebrick', 'label': 'hottest removed'},
    'mean_diff_trim_cold': {'colour': 'steelblue', 'label': 'coldest removed'},
}

N_CONNECTORS = 12  # rank matched pairs drawn on the method figure


def tidy(ax):
    """
    Drop the top and right spines so the data is not boxed in.
    Parameters
    ----------
    ax (matplotlib.axes.Axes) :Axes to tidy.
    Returns
    -------
    ax (matplotlib.axes.Axes) :The same axes.
    """
    ax.spines[['top', 'right']].set_visible(False)
    return ax


def plot_method(averaged, season='DJF', axes=None):
    """
    Three panels walking through the method on one season. The first shows the
    two quantile functions with a handful of rank matched pairs joined, which
    is the step that makes two unpaired ensembles comparable: the i-th coldest
    day of one is set against the i-th coldest of the other, not against the
    same calendar day. The second shows those differences, which by the sorting
    identity average to exactly the difference in means, so the panel is the
    mean shift taken apart by where in the distribution it comes from. The
    third shows the mean shift remaining as days are deleted from each end,
    which is the original framing of the question and the cumulative integral
    of the second panel.
    Parameters
    ----------
    averaged (xr.Dataset) :Decomposition averaged over season_year, carrying the quantile functions and both trimmed mean curves.
    season (str) :Season to illustrate.
    axes (array of matplotlib.axes.Axes, optional) :Three axes to draw into, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    one = averaged.sel(season=season)
    if axes is None:
        _, axes = plt.subplots(1, 3, figsize=(14, 4.2))

    quantile = one['quantile_level']
    connectors = np.linspace(0, one.sizes['rank'] - 1, N_CONNECTORS).astype(int)
    for rank in connectors:
        axes[0].plot(
            [quantile[rank]] * 2,
            [one.histnat_q[rank], one.hist_q[rank]],
            color='0.6', lw=0.8, zorder=1,
        )
    axes[0].plot(quantile, one.histnat_q, color='steelblue', lw=1.6, label='histnat', zorder=2)
    axes[0].plot(quantile, one.hist_q, color='firebrick', lw=1.6, label='hist', zorder=2)
    axes[0].set_xlabel('quantile')
    axes[0].set_ylabel('temperature (degC)')
    axes[0].set_title('1. match rank for rank', loc='left')
    axes[0].legend(frameon=False, loc='upper left')

    axes[1].fill_between(quantile, 0, one.quantile_diff, color='firebrick', alpha=0.18)
    axes[1].plot(quantile, one.quantile_diff, color='firebrick', lw=1.2)
    axes[1].axhline(float(one.mean_diff), color='k', lw=0.9, ls='--')
    axes[1].annotate(
        f'mean shift {float(one.mean_diff):.2f} K',
        xy=(0.03, float(one.mean_diff)), xytext=(0, 6),
        textcoords='offset points', fontsize=9,
    )
    axes[1].set_xlabel('quantile')
    axes[1].set_ylabel('hist - histnat (K)')
    axes[1].set_title('2. difference at each rank', loc='left')

    for name, style in TRIM_COLOURS.items():
        axes[2].plot(
            one.n_removed, one[name], color=style['colour'], lw=1.2,
            label=style['label'],
        )
    axes[2].axhline(float(one.mean_diff), color='k', lw=0.9, ls='--')
    axes[2].axhline(0, color='k', lw=0.7)
    axes[2].set_xlabel('days removed')
    axes[2].set_ylabel('remaining mean shift (K)')
    axes[2].set_title('3. trim from each end', loc='left')
    axes[2].legend(frameon=False)

    for ax in axes:
        tidy(ax)
    return axes


def plot_results(averaged, results, axes=None):
    """
    Four panels carrying the result. The top row is distribution shape averaged
    over recent seasons, since a single season is far too noisy to read shape
    from, and the bottom row is the full record. Warming against quantile
    flat on its dashed reference means the distribution translated, the share
    curve on the diagonal means extremes contributed exactly their proportion
    of the mean change and no more, and the standard deviation ratio flat at
    one means the spread never changed. Those three say the same thing three
    independent ways, which is what makes the conclusion defensible.
    Parameters
    ----------
    averaged (xr.Dataset) :Decomposition averaged over season_year, carrying quantile_diff, mean_diff and share.
    results (xr.Dataset) :Full record decomposition, carrying mean_diff and qq_slope.
    axes (array of matplotlib.axes.Axes, optional) :Four axes to draw into, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    if axes is None:
        _, axes = plt.subplots(2, 2, figsize=(12, 8.5))
    axes = np.asarray(axes).ravel()

    for season, colour in SEASON_COLOURS.items():
        shape = averaged.sel(season=season)
        axes[0].plot(shape['quantile_level'], shape.quantile_diff, color=colour, lw=1.1, label=season)
        axes[0].axhline(float(shape.mean_diff), color=colour, lw=0.7, ls='--')
        axes[1].plot(shape.hottest_fraction, shape.share, color=colour, lw=1.4, label=season)
        axes[2].plot(results.season_year, results.mean_diff.sel(season=season), color=colour, lw=0.9, label=season)
        axes[3].plot(results.season_year, results.qq_slope.sel(season=season), color=colour, lw=0.9, label=season)

    axes[0].set_xlabel('quantile')
    axes[0].set_ylabel('hist - histnat (K)')
    axes[0].set_title('warming is the same at every quantile', loc='left')
    axes[0].legend(frameon=False, ncol=4, fontsize=9)

    axes[1].plot([0, 1], [0, 1], color='k', lw=0.9, ls='--')
    axes[1].annotate(
        'equal share', xy=(0.62, 0.62), xytext=(6, -14),
        textcoords='offset points', rotation=45, fontsize=9, color='0.3',
    )
    axes[1].set_xlabel('hottest fraction of days')
    axes[1].set_ylabel('share of mean shift')
    axes[1].set_title('extremes carry only their proportion', loc='left')

    axes[2].axhline(0, color='k', lw=0.7)
    axes[2].set_xlabel('season year')
    axes[2].set_ylabel('mean shift (K)')
    axes[2].set_title('when the shift emerges', loc='left')

    axes[3].axhline(1, color='k', lw=0.9, ls='--')
    axes[3].set_xlabel('season year')
    axes[3].set_ylabel('hist / histnat standard deviation')
    axes[3].set_title('spread is unchanged', loc='left')

    for ax in axes:
        tidy(ax)
    return axes


def plot_bands(results, axes=None):
    """
    Cold tail, middle and hot tail mean differences through the record, one
    panel per season. Under a uniform shift the three track each other, and a
    band pulling systematically away from the others is where the distribution
    is changing shape. The cold tail is far noisier than the middle because it
    averages a tenth of the ranks in the sparsest part of the distribution, so
    its larger swings are not evidence of anything on their own.
    Parameters
    ----------
    results (xr.Dataset) :Decomposition carrying the band mean differences and a season dimension.
    axes (array of matplotlib.axes.Axes, optional) :One axes per season, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    if axes is None:
        _, axes = plt.subplots(
            len(SEASON_COLOURS), 1, figsize=(8, 10), sharex=True, sharey=True,
        )

    for ax, season in zip(axes, SEASON_COLOURS):
        one = results.sel(season=season)
        for name, colour in BAND_COLOURS.items():
            ax.plot(
                one.season_year, one[name], color=colour, lw=0.9,
                label=name.replace('diff_', '').replace('_', ' '),
            )
        ax.axhline(0, color='k', lw=0.7)
        ax.set_ylabel(f'{season} (K)')
        tidy(ax)

    axes[0].legend(frameon=False, ncol=3, fontsize=9)
    axes[-1].set_xlabel('season year')
    return axes

```

```python
"""
Rank decomposition of the forced mean shift, season by season.
Sorting does not change a mean, so for two samples of equal length
mean(hist) - mean(histnat) equals the mean over i of x_(i) - y_(i), where
x_(i) and y_(i) are the i-th smallest values of each sample. The mean shift
therefore partitions exactly across ranks, which is what makes quantile_diff
an exact decomposition rather than a suggestive picture, and what lets it say
where in the distribution the warming lives.
Season labelling follows the convention that a season is named for the year
its January falls in, so DJF 2003/04 has season_year 2004. Only DJF crosses a
year boundary, so only DJF takes that offset.
"""

import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

SEASONS = ['DJF', 'MAM', 'JJA', 'SON']
DAYS_PER_SEASON = 90  # 360 day calendar, so every season is exactly 90 steps
MIN_RETAINED = 30  # trimmed means over fewer ranks than this are unusable
N_QUANTILE_LEVELS = 1000  # grid size when not matching rank for rank

# Quantile ranges reduced to a single mean difference each. These are the only
# form of the decomposition that survives to a map, since 4050 ranks per
# gridpoint cannot be stored continent wide.
BANDS = {
    'cold_tail': {'lower': 0.0, 'upper': 0.1},
    'middle': {'lower': 0.25, 'upper': 0.75},
    'hot_tail': {'lower': 0.9, 'upper': 1.0},
}


# ---------------------------------------------------------------- preparation

def add_season_year(da, season):
    """
    Label each timestep with the year its season ends in. Only DJF crosses a
    year boundary, so only there does December need pushing forward to sit with
    the following January and February. Filtering on time.dt.year directly
    would otherwise mix the tail of one summer with the head of the next. For
    the other three seasons the label is just the calendar year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data with a cftime time coordinate.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    labelled (xr.DataArray) :Input with a season_year coordinate added along time.
    """
    rolls_over = (da.time.dt.month == 12) & (season == 'DJF')
    labelled = da.assign_coords(season_year=da.time.dt.year + rolls_over)
    return labelled


def full_season_years(da):
    """
    Find season years holding a complete season, used to drop the part seasons
    at each end of the record where the opening or closing months fall outside
    the data.
    Parameters
    ----------
    da (xr.DataArray) :Data carrying a season_year coordinate along time.
    Returns
    -------
    complete (np.ndarray) :Season years containing exactly DAYS_PER_SEASON timesteps.
    """
    counts = da.season_year.groupby('season_year').count()
    complete = counts.season_year.values[counts.values == DAYS_PER_SEASON]
    return complete


def match_dims(hist, histnat, season):
    """
    Reduce both experiments to one season over a common set of members and
    complete seasons. Rank matching is meaningless unless the two samples are
    the same size, so equal sample size is enforced here rather than assumed
    downstream, where a mismatch would silently compare offset ranks.
    Parameters
    ----------
    hist (xr.DataArray) :Daily temperature from the historical experiment, with member and time.
    histnat (xr.DataArray) :Daily temperature from the natural forcing only experiment.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    hist, histnat (xr.DataArray) :The two experiments subset to matching members and complete seasons, loaded into memory.
    """
    hist = add_season_year(
        hist.sel(time=hist.time.dt.season == season), season
    ).sortby('time')
    histnat = add_season_year(
        histnat.sel(time=histnat.time.dt.season == season), season
    ).sortby('time')

    # hist and histnat members share no initial conditions, so this buys equal
    # sample size for rank matching and nothing more. It is not physical pairing.
    members = np.intersect1d(hist.member, histnat.member)
    hist = hist.sel(member=members)
    histnat = histnat.sel(member=members)

    years = np.intersect1d(full_season_years(hist), full_season_years(histnat))
    hist = hist.sel(time=hist.season_year.isin(years)).load()
    histnat = histnat.sel(time=histnat.season_year.isin(years)).load()

    # to_year_day assigns day numbers by tiling, which is only valid if the
    # sorted record really is contiguous blocks of DAYS_PER_SEASON.
    expected = np.repeat(years, DAYS_PER_SEASON)
    assert (hist.season_year.values == expected).all()
    assert (histnat.season_year.values == expected).all()
    return hist, histnat


# ------------------------------------------------------------------- ranking

def to_year_day(da):
    """
    Reshape the flat time axis into separate season_year and day dimensions.
    Purely structural with no arithmetic, but it is what allows each season to
    be sorted independently in one vectorised call rather than through a
    groupby, which would build a separate dask graph per year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data carrying a season_year coordinate, as returned by prepare_pair.
    Returns
    -------
    reshaped (xr.DataArray) :Input with time replaced by season_year and day.
    """
    n_years = np.unique(da.season_year.values).size
    reshaped = (
        da.assign_coords(
            day=('time', np.tile(np.arange(DAYS_PER_SEASON), n_years))
        )
        .set_index(time=('season_year', 'day'))
        .unstack('time')
    )
    return reshaped


def sort_pooled(arr):
    """
    Pool the trailing two axes into one and sort it ascending. Operating on the
    last two axes is what lets apply_ufunc broadcast this over any leading
    dimensions, so the same kernel works at one gridpoint or across a map.
    Parameters
    ----------
    arr (np.ndarray) :Array whose final two axes are member and day.
    Returns
    -------
    ordered (np.ndarray) :Same leading axes, with the trailing two replaced by one sorted axis.
    """
    ordered = np.sort(arr.reshape(arr.shape[:-2] + (-1,)), axis=-1)
    return ordered


def rank_sort(da):
    """
    Build the empirical quantile function by pooling every member and every day
    into a single sample and sorting it. Members are treated as alternative
    realisations of the same season, so the resulting distribution describes
    the days that could have occurred in that year rather than the days one
    member happened to produce. Any dimensions other than member and day, such
    as season_year or season, are left alone and sorted independently.
    Parameters
    ----------
    da (xr.DataArray) :Data with member and day dimensions, as returned by to_year_day.
    Returns
    -------
    ranked (xr.DataArray) :Leading dimensions of da with member and day replaced by rank, ascending along rank, carrying a quantile_level coordinate giving the plotting position of each rank.
    """
    n_sample = da.sizes['member'] * da.sizes['day']

    # Naming member and day as core dims moves them to the trailing axes, which
    # is exactly what sort_pooled expects to flatten.
    ranked = xr.apply_ufunc(
        sort_pooled,
        da,
        input_core_dims=[['member', 'day']],
        output_core_dims=[['rank']],
    )

    # quantile_level, not quantile, because attribute access resolves methods
    # before coordinates and DataArray.quantile would shadow it.
    ranked = ranked.assign_coords(
        quantile_level=('rank', (np.arange(n_sample) + 0.5) / n_sample)
    )
    return ranked


def quantile_grid_function(da, pool_dims):
    """
    Empirical quantile function evaluated on a fixed grid of probability levels
    rather than by sorting. Sorting requires the two samples to be the same
    size, since it matches the i-th smallest of one against the i-th smallest
    of the other, and that constraint is what forces equal member counts and
    equal season lengths. Evaluating both sides on a shared probability grid
    lifts it, which is what allows records of unequal length, calendars where a
    season is sometimes 91 days, and a reference pooled over far more values
    than the sample it is compared against. The cost is that the decomposition
    becomes a very close approximation rather than an exact identity, because
    mean_diff is then a mean over grid points instead of over every value.
    Parameters
    ----------
    da (xr.DataArray) :Data with member and day dimensions, as returned by to_year_day.
    pool_dims (list of str) :Dimensions pooled before evaluating quantiles, member and day for a per season distribution, or member, day and season_year for a reference pooled over the whole record.
    Returns
    -------
    evaluated (xr.DataArray) :Dimensions of da less pool_dims, plus rank indexing the probability grid and carrying a quantile_level coordinate.
    """
    levels = (np.arange(N_QUANTILE_LEVELS) + 0.5) / N_QUANTILE_LEVELS

    # quantile names its new dimension after itself, so it is renamed to rank
    # to keep every downstream function unchanged. In this path rank indexes
    # probability levels rather than sample order.
    evaluated = (
        da.quantile(levels, dim=pool_dims)
        .rename(quantile='rank')
        .drop_vars('rank')
        .assign_coords(quantile_level=('rank', levels))
    )
    return evaluated


# ------------------------------------------------------- trimmed mean shifts

def trimmed_mean_diff(quantile_diff, from_hot_end):
    """
    Mean rank difference remaining after deleting a growing number of days from
    one end of the distribution. This is the cumulative form of the original
    question, which asked how many extreme events would have to be removed
    before the mean returns to the counterfactual. Both directions are offered
    because whichever tail is least changed is the wrong one to delete.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, with a rank dimension.
    from_hot_end (bool) :True deletes the hottest days first, False the coldest.
    Returns
    -------
    curve (xr.DataArray) :Leading dimensions of quantile_diff with rank replaced by n_removed, which counts deleted days and runs upwards from zero.
    """
    diff = quantile_diff.transpose(..., 'rank')

    # Ranks ascend, so a cumulative sum accumulates from the cold end and
    # therefore describes deleting the hottest days. Reversing the input first
    # flips which end is being deleted.
    values = diff.values if from_hot_end else diff.values[..., ::-1]

    n_rank = diff.sizes['rank']
    retained = np.arange(1, n_rank + 1)

    # cumsum is indexed by days kept, so both it and the divisor are reversed
    # to give an output indexed by days removed, ascending from zero.
    means = np.cumsum(values, axis=-1)[..., ::-1] / retained[::-1]

    curve = xr.DataArray(
        means,
        dims=diff.dims[:-1] + ('n_removed',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(n_removed=('n_removed', np.arange(n_rank)))
    return curve


def share_of_mean_shift(quantile_diff):
    """
    Fraction of the total mean shift carried by the hottest fraction of the
    distribution, which is the statistic the whole diagnostic was built to
    answer. Under a uniform shift the hottest tenth of days carries a tenth of
    the mean change, so the curve lies on the diagonal, and extremes carrying
    more than their share bows it above. This asks about proportion rather than
    about a threshold being reached, so unlike a crossing point it has an
    answer for every season rather than only for those where trimming happens
    to cancel the shift. It divides by the total shift, so it is meaningless in
    seasons where that total is near zero and the forced signal has not
    emerged. Rank differences can be negative, so unlike the precipitation
    version of this diagnostic the share is not confined between zero and one,
    and a value above one means part of the distribution cooled.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, with a rank dimension.
    Returns
    -------
    share (xr.DataArray) :Leading dimensions of quantile_diff with rank replaced by hottest_fraction, running from one rank up to the whole distribution.
    """
    diff = quantile_diff.transpose(..., 'rank')
    n_rank = diff.sizes['rank']

    # Ranks ascend, so the cumulative sum is reversed to accumulate from the
    # hot end. The final element is the total shift, which normalises the rest.
    cumulative = np.cumsum(diff.values[..., ::-1], axis=-1)
    share = cumulative / cumulative[..., -1:]

    share = xr.DataArray(
        share,
        dims=diff.dims[:-1] + ('hottest_fraction',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(
        hottest_fraction=('hottest_fraction', np.arange(1, n_rank + 1) / n_rank)
    )
    return share

```

```python
def match_dims_by_season(da1, da2, seasons=SEASONS):
    """
    Run match_dims for each season and stack the results on a season dimension,
    giving one array per record with season, season_year, member and day. The
    decomposition never names season, so once the records are shaped this way
    every season is decomposed independently with no further looping.
    Parameters
    ----------
    da1 (xr.DataArray) :Daily temperature with member and time, conventionally the forced record.
    da2 (xr.DataArray) :Daily temperature to compare against, on the same dimensions.
    seasons (list of str) :Season names to include, defaulting to all four.
    Returns
    -------
    da1_prepared, da2_prepared (xr.DataArray) :The two records with dimensions season, season_year, member and day.
    """
    da1_by_season = []
    da2_by_season = []
    for season in seasons:
        da1_season, da2_season = match_dims(da1, da2, season)
        da1_by_season.append(to_year_day(da1_season))
        da2_by_season.append(to_year_day(da2_season))
 
    # join='inner' keeps only the season years every season has, since DJF
    # loses one at each end. Outer joining would pad the others with NaN, and
    # NaN sorts to the hot end of the distribution and would corrupt the
    # ranking rather than just leaving a gap.
    season_dim = xr.DataArray(seasons, dims='season', name='season')
    da1_prepared = xr.concat(da1_by_season, dim=season_dim, join='inner')
    da2_prepared = xr.concat(da2_by_season, dim=season_dim, join='inner')
    return da1_prepared, da2_prepared
 
```

```python

def rank_decomposition_full_set(da1, da2, use_quantile_grid=False,
                                pooled_reference=False):
    """
    Run the decomposition and add every summary of it that the figures use, so
    that one call produces a complete results object. rank_decomposition stays
    separate and returns only the exact decomposition, since the summaries
    below are choices rather than consequences, and dropping one should be a
    matter of deleting a line here rather than editing the core.
    Parameters
    ----------
    da1 (xr.DataArray) :Prepared data with member and day dimensions, conventionally the forced record.
    da2 (xr.DataArray) :Prepared data to compare against, on the same dimensions.
    use_quantile_grid (bool) :Passed to rank_decomposition.
    pooled_reference (bool) :Passed to rank_decomposition.
    Returns
    -------
    full_set (xr.Dataset) :The decomposition together with both trimmed mean curves, the share of the mean shift, the standard deviation ratio and the band mean differences.
    """
    results = rank_decomposition(
        da1,
        da2,
        use_quantile_grid=use_quantile_grid,
        pooled_reference=pooled_reference,
    )
    quantile_diff = results.quantile_diff
 
    mean_diff_trim_hot = trimmed_mean_diff(quantile_diff, from_hot_end=True)
    mean_diff_trim_cold = trimmed_mean_diff(quantile_diff, from_hot_end=False)
    share = share_of_mean_shift(quantile_diff)
    band_diffs = band_mean_diffs(quantile_diff)
 
    # The intercept is the fit extrapolated to zero degrees, tens of degrees
    # outside polar data, so it is discarded and mean_diff carries the shift.
    qq_slope, _ = quantile_quantile_fit(results.hist_q, results.histnat_q)
 
    full_set = results.assign(
        mean_diff_trim_hot=mean_diff_trim_hot,
        mean_diff_trim_cold=mean_diff_trim_cold,
        share=share,
        qq_slope=qq_slope,
        **band_diffs,
    )
    return full_set
 
```

```python
"""
Rank decomposition of the forced mean shift, season by season.
Sorting does no`\t change a mean, so for two samples of equal length
mean(hist) - mean(histnat) equals the mean over i of x_(i) - y_(i), where
x_(i) and y_(i) are the i-th smallest values of each sample. The mean shift
therefore partitions exactly across ranks, which is what makes quantile_diff
an exact decomposition rather than a suggestive picture, and what lets it say
where in the distribution the warming lives.
Season labelling follows the convention that a season is named for the year
its January falls in, so DJF 2003/04 has season_year 2004. Only DJF crosses a
year boundary, so only DJF takes that offset.
"""

import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

SEASONS = ['DJF', 'MAM', 'JJA', 'SON']
DAYS_PER_SEASON = 90  # 360 day calendar, so every season is exactly 90 steps
MIN_RETAINED = 30  # trimmed means over fewer ranks than this are unusable
N_QUANTILE_LEVELS = 1000  # grid size when not matching rank for rank

# Quantile ranges reduced to a single mean difference each. These are the only
# form of the decomposition that survives to a map, since 4050 ranks per
# gridpoint cannot be stored continent wide.
BANDS = {
    'cold_tail': {'lower': 0.0, 'upper': 0.1},
    'middle': {'lower': 0.25, 'upper': 0.75},
    'hot_tail': {'lower': 0.9, 'upper': 1.0},
}


# ---------------------------------------------------------------- preparation

def add_season_year(da, season):
    """
    Label each timestep with the year its season ends in. Only DJF crosses a
    year boundary, so only there does December need pushing forward to sit with
    the following January and February. Filtering on time.dt.year directly
    would otherwise mix the tail of one summer with the head of the next. For
    the other three seasons the label is just the calendar year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data with a cftime time coordinate.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    labelled (xr.DataArray) :Input with a season_year coordinate added along time.
    """
    rolls_over = (da.time.dt.month == 12) & (season == 'DJF')
    labelled = da.assign_coords(season_year=da.time.dt.year + rolls_over)
    return labelled


def full_season_years(da):
    """
    Find season years holding a complete season, used to drop the part seasons
    at each end of the record where the opening or closing months fall outside
    the data.
    Parameters
    ----------
    da (xr.DataArray) :Data carrying a season_year coordinate along time.
    Returns
    -------
    complete (np.ndarray) :Season years containing exactly DAYS_PER_SEASON timesteps.
    """
    counts = da.season_year.groupby('season_year').count()
    complete = counts.season_year.values[counts.values == DAYS_PER_SEASON]
    return complete


def prepare_pair(hist, histnat, season):
    """
    Reduce both experiments to one season over a common set of members and
    complete seasons. Rank matching is meaningless unless the two samples are
    the same size, so equal sample size is enforced here rather than assumed
    downstream, where a mismatch would silently compare offset ranks.
    Parameters
    ----------
    hist (xr.DataArray) :Daily temperature from the historical experiment, with member and time.
    histnat (xr.DataArray) :Daily temperature from the natural forcing only experiment.
    season (str) :Three letter season name, one of SEASONS.
    Returns
    -------
    hist, histnat (xr.DataArray) :The two experiments subset to matching members and complete seasons, loaded into memory.
    """
    hist = add_season_year(
        hist.sel(time=hist.time.dt.season == season), season
    ).sortby('time')
    histnat = add_season_year(
        histnat.sel(time=histnat.time.dt.season == season), season
    ).sortby('time')

    # hist and histnat members share no initial conditions, so this buys equal
    # sample size for rank matching and nothing more. It is not physical pairing.
    members = np.intersect1d(hist.member, histnat.member)
    hist = hist.sel(member=members)
    histnat = histnat.sel(member=members)

    years = np.intersect1d(full_season_years(hist), full_season_years(histnat))
    hist = hist.sel(time=hist.season_year.isin(years)).load()
    histnat = histnat.sel(time=histnat.season_year.isin(years)).load()

    # to_year_day assigns day numbers by tiling, which is only valid if the
    # sorted record really is contiguous blocks of DAYS_PER_SEASON.
    expected = np.repeat(years, DAYS_PER_SEASON)
    assert (hist.season_year.values == expected).all()
    assert (histnat.season_year.values == expected).all()
    return hist, histnat


def prepare_seasons(hist, histnat, seasons=SEASONS):
    """
    Run prepare_pair for several seasons and collect the results, so the
    matching decisions for each season stay visible as their own object before
    anything is decomposed. Each season is matched independently, so the
    surviving years can differ between them, which is expected since only DJF
    loses a year at each end of the record.
    Parameters
    ----------
    hist (xr.DataArray) :Daily temperature from the historical experiment, with member and time.
    histnat (xr.DataArray) :Daily temperature from the natural forcing only experiment.
    seasons (list of str) :Season names to prepare, defaulting to all four.
    Returns
    -------
    prepared (dict of str to dict) :Season name mapped to a dict with hist and histnat keys.
    """
    prepared = {}
    for season in seasons:
        hist_season, histnat_season = prepare_pair(hist, histnat, season)
        prepared[season] = {'hist': hist_season, 'histnat': histnat_season}
    return prepared


# ------------------------------------------------------------------- ranking

def to_year_day(da):
    """
    Reshape the flat time axis into separate season_year and day dimensions.
    Purely structural with no arithmetic, but it is what allows each season to
    be sorted independently in one vectorised call rather than through a
    groupby, which would build a separate dask graph per year.
    Parameters
    ----------
    da (xr.DataArray) :Daily data carrying a season_year coordinate, as returned by prepare_pair.
    Returns
    -------
    reshaped (xr.DataArray) :Input with time replaced by season_year and day.
    """
    n_years = np.unique(da.season_year.values).size
    reshaped = (
        da.assign_coords(
            day=('time', np.tile(np.arange(DAYS_PER_SEASON), n_years))
        )
        .set_index(time=('season_year', 'day'))
        .unstack('time')
    )
    return reshaped


def sort_pooled(arr):
    """
    Pool the trailing two axes into one and sort it ascending. Operating on the
    last two axes is what lets apply_ufunc broadcast this over any leading
    dimensions, so the same kernel works at one gridpoint or across a map.
    Parameters
    ----------
    arr (np.ndarray) :Array whose final two axes are member and day.
    Returns
    -------
    ordered (np.ndarray) :Same leading axes, with the trailing two replaced by one sorted axis.
    """
    ordered = np.sort(arr.reshape(arr.shape[:-2] + (-1,)), axis=-1)
    return ordered


def rank_sort(da):
    """
    Build the per season empirical quantile function by pooling every member
    and every day into a single sample and sorting it. Members are treated as
    alternative realisations of the same season, so the resulting distribution
    describes the days that could have occurred in that year rather than the
    days one member happened to produce.
    Parameters
    ----------
    da (xr.DataArray) :Daily data with member and time, as returned by prepare_pair.
    Returns
    -------
    ranked (xr.DataArray) :Dimensions season_year by rank, ascending along rank, carrying a quantile_level coordinate giving the plotting position of each rank.
    """
    da = to_year_day(da)
    n_sample = da.sizes['member'] * da.sizes['day']

    # Naming member and day as core dims moves them to the trailing axes, which
    # is exactly what sort_pooled expects to flatten.
    ranked = xr.apply_ufunc(
        sort_pooled,
        da,
        input_core_dims=[['member', 'day']],
        output_core_dims=[['rank']],
    )

    # quantile_level, not quantile, because attribute access resolves methods
    # before coordinates and DataArray.quantile would shadow it.
    ranked = ranked.assign_coords(
        quantile_level=('rank', (np.arange(n_sample) + 0.5) / n_sample)
    )
    return ranked


def quantile_grid_function(da, pool_dims):
    """
    Empirical quantile function evaluated on a fixed grid of probability levels
    rather than by sorting. Sorting requires the two samples to be the same
    size, since it matches the i-th smallest of one against the i-th smallest
    of the other, and that constraint is what forces equal member counts and
    equal season lengths. Evaluating both sides on a shared probability grid
    lifts it, which is what allows records of unequal length, calendars where a
    season is sometimes 91 days, and a reference pooled over far more values
    than the sample it is compared against. The cost is that the decomposition
    becomes a very close approximation rather than an exact identity, because
    mean_diff is then a mean over grid points instead of over every value.
    Parameters
    ----------
    da (xr.DataArray) :Daily data carrying a season_year coordinate, as returned by prepare_pair.
    pool_dims (list of str) :Dimensions pooled before evaluating quantiles, member and day for a per season distribution, or member, day and season_year for a reference pooled over the whole record.
    Returns
    -------
    evaluated (xr.DataArray) :Dimensions of da less pool_dims, plus rank indexing the probability grid and carrying a quantile_level coordinate.
    """
    levels = (np.arange(N_QUANTILE_LEVELS) + 0.5) / N_QUANTILE_LEVELS

    # quantile names its new dimension after itself, so it is renamed to rank
    # to keep every downstream function unchanged. In this path rank indexes
    # probability levels rather than sample order.
    evaluated = (
        to_year_day(da)
        .quantile(levels, dim=pool_dims)
        .rename(quantile='rank')
        .drop_vars('rank')
        .assign_coords(quantile_level=('rank', levels))
    )
    return evaluated


# ------------------------------------------------------- trimmed mean shifts

def trimmed_mean_diff(quantile_diff, from_hot_end):
    """
    Mean rank difference remaining after deleting a growing number of days from
    one end of the distribution. This is the cumulative form of the original
    question, which asked how many extreme events would have to be removed
    before the mean returns to the counterfactual. Both directions are offered
    because whichever tail is least changed is the wrong one to delete.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, with a rank dimension.
    from_hot_end (bool) :True deletes the hottest days first, False the coldest.
    Returns
    -------
    curve (xr.DataArray) :Leading dimensions of quantile_diff with rank replaced by n_removed, which counts deleted days and runs upwards from zero.
    """
    diff = quantile_diff.transpose(..., 'rank')

    # Ranks ascend, so a cumulative sum accumulates from the cold end and
    # therefore describes deleting the hottest days. Reversing the input first
    # flips which end is being deleted.
    values = diff.values if from_hot_end else diff.values[..., ::-1]

    n_rank = diff.sizes['rank']
    retained = np.arange(1, n_rank + 1)

    # cumsum is indexed by days kept, so both it and the divisor are reversed
    # to give an output indexed by days removed, ascending from zero.
    means = np.cumsum(values, axis=-1)[..., ::-1] / retained[::-1]

    curve = xr.DataArray(
        means,
        dims=diff.dims[:-1] + ('n_removed',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(n_removed=('n_removed', np.arange(n_rank)))
    return curve


def days_to_crossing(curve):
    """
    Number of days that must be deleted before the trimmed mean difference
    changes sign against the untrimmed shift, at which point trimming has
    removed the warming entirely. NaN means no amount of trimming recovers the
    counterfactual mean, which is the expected outcome wherever the shift is
    uniform. Note that requiring a sign change is both too strict and too
    loose: a curve that decays to zero and flattens never crosses, while a
    curve sitting near zero crosses on noise alone, so this fires most readily
    in seasons where there is no signal to detect.
    Parameters
    ----------
    curve (xr.DataArray) :Trimmed mean difference with an n_removed dimension, from trimmed_mean_diff.
    Returns
    -------
    crossing (xr.DataArray) :Leading dimensions of curve, holding the crossing point in days, or NaN where the curve never crosses.
    """
    # Past this point the trimmed mean is taken over a handful of ranks and its
    # sampling error swamps any signal, so crossings there are meaningless.
    usable = curve.n_removed <= curve.sizes['n_removed'] - MIN_RETAINED
    crossed = (np.sign(curve) != np.sign(curve.isel(n_removed=0))) & usable

    # argmax gives the first True, but returns zero when nothing is True, which
    # would read as an immediate crossing rather than none at all.
    crossing = xr.where(
        crossed.any('n_removed'),
        crossed.argmax('n_removed'),
        np.nan,
    )
    return crossing


# ------------------------------------------------------------ distribution fit

def quantile_quantile_fit(hist_q, histnat_q):
    """
    Least squares fit of the hist quantile function against the histnat
    quantile function, which separates a change in location from a change in
    spread. Writing x for histnat_q and y for hist_q along rank, the slope is
    sum((x_i - xbar)(y_i - ybar)) / sum((x_i - xbar)^2) and the intercept is
    ybar - slope * xbar, which is a degree one polyfit expressed as covariance
    over variance so that it vectorises over the leading dimensions without a
    Python loop. Because both inputs are sorted this is the quantile quantile
    fit, so the slope is effectively the ratio of standard deviations: one
    means the distribution simply translated and every quantile warmed by the
    same amount, above one means it stretched so extremes warmed more, below
    one means it compressed.
    The intercept is the fitted value at zero degrees, which for polar data is
    tens of degrees outside the sample, so small slope errors are levered into
    large intercept errors. Use mean_diff for the location shift, not this.
    Parameters
    ----------
    hist_q (xr.DataArray) :Sorted quantile function of the historical experiment, with a rank dimension, as returned by rank_sort.
    histnat_q (xr.DataArray) :Sorted quantile function of the natural forcing only experiment.
    Returns
    -------
    slope, intercept (xr.DataArray) :Fit coefficients over the leading dimensions of the inputs.
    """
    histnat_q_centred = histnat_q - histnat_q.mean('rank')
    slope = (
        histnat_q_centred * (hist_q - hist_q.mean('rank'))
    ).sum('rank') / (histnat_q_centred ** 2).sum('rank')
    intercept = hist_q.mean('rank') - slope * histnat_q.mean('rank')
    return slope, intercept


def band_mean_diffs(quantile_diff):
    """
    Mean rank difference within each quantile range in BANDS, reducing the
    decomposition from thousands of ranks to one number per band. These are the
    only summaries compact enough to carry to a map, and comparing them across
    seasons is how a uniform shift shows itself, since under translation the
    tails and the middle track each other. The bands cover different numbers of
    ranks in regions of different density, so their sampling errors differ by
    several times and the curves are not comparable at face value.
    Parameters
    ----------
    quantile_diff (xr.DataArray) :Rank differences of hist minus histnat, carrying a quantile_level coordinate along rank.
    Returns
    -------
    diffs (dict of str to xr.DataArray) :Band name prefixed with diff_, mapped to the mean difference over that band with the rank dimension reduced away.
    """
    level = quantile_diff['quantile_level']

    # where masks to the band and mean skips the resulting NaN, so each band is
    # averaged over its own ranks only.
    diffs = {}
    for name, band in BANDS.items():
        within = (level >= band['lower']) & (level <= band['upper'])
        diffs[f'diff_{name}'] = quantile_diff.where(within).mean('rank')
    return diffs


# --------------------------------------------------------------- decomposition

def rank_decomposition(hist_season, histnat_season, use_quantile_grid=False,
                       pooled_reference=False):
    """
    Decompose the hist minus histnat mean shift across the distribution for one
    prepared season. This returns only the decomposition itself, since
    quantile_diff and mean_diff are exact consequences of the sorting identity
    whereas every summary of them is a choice. The trimmed mean curves, the
    crossing points, the quantile quantile fit and the band means are separate
    functions to be added explicitly, so that what a given results object
    contains is visible at the call site rather than buried here.
    Parameters
    ----------
    hist_season (xr.DataArray) :Prepared daily data for one season from the historical experiment.
    histnat_season (xr.DataArray) :Prepared daily data for one season from the natural forcing only experiment.
    use_quantile_grid (bool) :Evaluate both distributions on a shared probability grid instead of matching rank for rank, which allows the two samples to differ in size at the cost of the decomposition becoming approximate.
    pooled_reference (bool) :Compare each season against histnat pooled over its whole record rather than against the matching histnat season, which removes the sampling noise contributed by the reference. Requires use_quantile_grid since the pooled sample is far larger.
    Returns
    -------
    results (xr.Dataset) :The two quantile functions, their rank difference and the mean shift it sums to.
    """
    assert use_quantile_grid or not pooled_reference, (
        'a pooled reference holds many more values than one hist season, so '
        'rank matching cannot align them. Set use_quantile_grid=True.'
    )

    # Pooling season_year into the reference collapses it, so histnat_q loses
    # its season dimension and broadcasts against every hist season.
    reference_dims = ['member', 'day']
    if pooled_reference:
        reference_dims = reference_dims + ['season_year']

    if use_quantile_grid:
        hist_q = quantile_grid_function(hist_season, ['member', 'day'])
        histnat_q = quantile_grid_function(histnat_season, reference_dims)
    else:
        hist_q = rank_sort(hist_season)
        histnat_q = rank_sort(histnat_season)

    # The i-th coldest hist day minus the i-th coldest histnat day. Not the
    # same days, the same positions in each distribution.
    quantile_diff = hist_q - histnat_q

    # Exactly mean(hist) - mean(histnat) by the sorting identity, so this is
    # the quantity trimming is trying to eliminate.
    mean_diff = quantile_diff.mean('rank')

    results = xr.Dataset({
        # The quantile functions are kept rather than discarded because a 1 K
        # difference means something different at -45 degC than at -10 degC,
        # and only the absolute curves say which.
        'hist_q': hist_q,
        'histnat_q': histnat_q,
        'quantile_diff': quantile_diff,
        'mean_diff': mean_diff,
    })
    return results


def decompose_seasons(prepared, use_quantile_grid=False, pooled_reference=False):
    """
    Run rank_decomposition on several already prepared seasons and stack them
    on a season dimension, so the whole annual cycle is one object rather than
    four separate runs. Preparation is deliberately not done here, since which
    members and which years survive the matching is a decision worth seeing
    rather than inheriting. Seasons are aligned on season_year when stacked,
    and DJF loses a year at each end of the record where the December or the
    following January and February fall outside the data, so those cells come
    back as NaN rather than silently shifting DJF against the other seasons.
    On a real calendar the seasons differ in length, which changes the size of
    the rank dimension and makes the stack fail, so use_quantile_grid is needed
    for anything that is not a 360 day calendar.
    Parameters
    ----------
    prepared (dict of str to dict) :Season name mapped to a dict with hist and histnat keys, as returned by prepare_seasons.
    use_quantile_grid (bool) :Passed to rank_decomposition.
    pooled_reference (bool) :Passed to rank_decomposition.
    Returns
    -------
    results (xr.Dataset) :Output of rank_decomposition for every season, stacked on a season dimension.
    """
    per_season = [
        rank_decomposition(
            pair['hist'],
            pair['histnat'],
            use_quantile_grid=use_quantile_grid,
            pooled_reference=pooled_reference,
        )
        for pair in prepared.values()
    ]

    results = xr.concat(
        per_season,
        dim=xr.DataArray(list(prepared), dims='season', name='season'),
    )
    return results


# ------------------------------------------------------------------- plotting

def plot_season(results, season_year, axes=None):
    """
    Trimmed mean curves and the rank difference curve for a single season. The
    right panel is the primary one, since the left is its cumulative integral
    and smears local structure across everything downstream of it.
    Parameters
    ----------
    results (xr.Dataset) :Output of rank_decomposition.
    season_year (int) :Season to draw.
    axes (array of matplotlib.axes.Axes, optional) :Pair of axes to draw into, created if not given.
    Returns
    -------
    axes (array of matplotlib.axes.Axes) :The axes drawn into.
    """
    one = results.sel(season_year=season_year)
    if axes is None:
        _, axes = plt.subplots(1, 2, figsize=(11, 4))

    curves = {
        'hottest removed': {
            'data': one.mean_diff_trim_hot,
            'colour': 'firebrick',
        },
        'coldest removed': {
            'data': one.mean_diff_trim_cold,
            'colour': 'steelblue',
        },
    }
    for label, curve in curves.items():
        axes[0].plot(
            curve['data'].n_removed,
            curve['data'],
            color=curve['colour'],
            label=label,
        )

    # Left panel stays in kelvin rather than normalised by mean_diff, which
    # blows up in early seasons where the forced signal has not emerged.
    axes[0].axhline(0, color='k', lw=0.6)
    axes[0].axhline(float(one.mean_diff), color='k', lw=0.6, ls='--')
    axes[0].set_xlabel('days removed')
    axes[0].set_ylabel('trimmed mean difference (K)')
    axes[0].set_title(str(int(season_year)))
    axes[0].legend(frameon=False)

    axes[1].plot(one['quantile_level'], one.quantile_diff, color='k', lw=0.8)
    axes[1].axhline(float(one.mean_diff), color='k', lw=0.6, ls='--')
    axes[1].set_xlabel('quantile')
    axes[1].set_ylabel('hist - histnat (K)')
    return axes


def plot_record(results, ax=None):
    """
    Every season at once, quantile against season year with the rank difference
    in colour. Horizontal banding means a uniform shift whose size varies year
    to year, vertical structure means the shift is concentrated in part of the
    distribution.
    Parameters
    ----------
    results (xr.Dataset) :Output of rank_decomposition.
    ax (matplotlib.axes.Axes, optional) :Axes to draw into, created if not given.
    Returns
    -------
    ax (matplotlib.axes.Axes) :The axes drawn into.
    """
    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))

    # robust clips the colour scale at the 2nd and 98th percentiles, which
    # otherwise saturate on the emergent decades and hide within season
    # structure entirely.
    results.quantile_diff.plot(
        x='quantile_level', y='season_year', center=0, robust=True, ax=ax
    )
    ax.set_xlabel('quantile')
    ax.set_ylabel('season year')
    return ax


def plot_quantile_functions(results, season_year, ax=None):
    """
    Absolute quantile functions for one season, for checking whether the hot
    tail is pinned against the melt point in both experiments. Only informative
    where the distribution actually reaches melting, and the offset between the
    curves is far smaller than the axis range wherever it does not.
    Parameters
    ----------
    results (xr.Dataset) :Output of rank_decomposition.
    season_year (int) :Season to draw.
    ax (matplotlib.axes.Axes, optional) :Axes to draw into, created if not given.
    Returns
    -------
    ax (matplotlib.axes.Axes) :The axes drawn into.
    """
    one = results.sel(season_year=season_year)
    if ax is None:
        _, ax = plt.subplots(figsize=(6, 4))

    experiments = {
        'hist': {'data': one.hist_q, 'colour': 'firebrick'},
        'histnat': {'data': one.histnat_q, 'colour': 'steelblue'},
    }
    for label, experiment in experiments.items():
        ax.plot(
            one['quantile_level'],
            experiment['data'],
            color=experiment['colour'],
            label=label,
        )

    ax.axhline(0.0, color='k', lw=0.6, ls=':')  # melt point, data in degC
    ax.set_xlabel('quantile')
    ax.set_ylabel('temperature (degC)')
    ax.legend(frameon=False)
    return ax


# --------------------------------------------------------------------- driver

prepared = prepare_seasons(
    hist_ensemble_ds.isel(lat=0, lon=0),
    histnat_ensemble_ds.isel(lat=0, lon=0),
)
results = decompose_seasons(prepared)

# Indices are added here rather than inside rank_decomposition, so that what
# results holds is whatever is listed below and nothing else.
results['mean_diff_trim_hot'] = trimmed_mean_diff(
    results.quantile_diff, from_hot_end=True
)
results['mean_diff_trim_cold'] = trimmed_mean_diff(
    results.quantile_diff, from_hot_end=False
)
results['days_to_crossing_hot'] = days_to_crossing(results.mean_diff_trim_hot)
results['days_to_crossing_cold'] = days_to_crossing(results.mean_diff_trim_cold)
results['qq_slope'], results['qq_intercept'] = quantile_quantile_fit(
    results.hist_q, results.histnat_q
)
results.update(band_mean_diffs(results.quantile_diff))

# The plot functions expect a single season, so pick one off the season dim.
djf = results.sel(season='DJF')
plot_record(djf)
plot_season(djf, djf.season_year.values[-1])
plot_quantile_functions(djf, djf.season_year.values[-1])
```

```python
"""Rank-decomposition of the forced mean shift, season by season.

Sorting does not change a mean, so for equal-length samples

    mean(hist) - mean(histnat) = (1/n) * sum_i (x_(i) - y_(i))

where x_(i), y_(i) are the i-th smallest values of each sample. The mean shift
therefore decomposes exactly across ranks, and `quantile_diff` says where in the
distribution the shift comes from.

Season labelling: a season is labelled by the year its January falls in, so
DJF 2003/04 has season_year 2004.
"""

import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

DAYS_PER_SEASON = 90  # 360-day calendar
MIN_RETAINED = 30  # trimmed means over fewer ranks than this are unusable

BANDS = {
    'cold_tail': {'lower': 0.0, 'upper': 0.1},
    'middle': {'lower': 0.25, 'upper': 0.75},
    'hot_tail': {'lower': 0.9, 'upper': 1.0},
}


# ---------------------------------------------------------------- preparation

def add_season_year(da):
    return da.assign_coords(
        season_year=da.time.dt.year + (da.time.dt.month == 12)
    )


def full_season_years(da):
    counts = da.season_year.groupby('season_year').count()
    return counts.season_year.values[counts.values == DAYS_PER_SEASON]


def prepare_pair(hist, histnat):
    """Select DJF, match members, drop part-seasons, load."""
    hist = add_season_year(
        hist.sel(time=hist.time.dt.season == 'DJF')
    ).sortby('time')
    histnat = add_season_year(
        histnat.sel(time=histnat.time.dt.season == 'DJF')
    ).sortby('time')

    members = np.intersect1d(hist.member, histnat.member)
    hist = hist.sel(member=members)
    histnat = histnat.sel(member=members)

    years = np.intersect1d(full_season_years(hist), full_season_years(histnat))
    hist = hist.sel(time=hist.season_year.isin(years)).load()
    histnat = histnat.sel(time=histnat.season_year.isin(years)).load()

    expected = np.repeat(years, DAYS_PER_SEASON)
    assert (hist.season_year.values == expected).all()
    assert (histnat.season_year.values == expected).all()
    return hist, histnat


# ------------------------------------------------------------------- ranking

def to_year_day(da):
    n_years = np.unique(da.season_year.values).size
    return (
        da.assign_coords(
            day=('time', np.tile(np.arange(DAYS_PER_SEASON), n_years))
        )
        .set_index(time=('season_year', 'day'))
        .unstack('time')
    )


def sort_pooled(arr):
    """Pool the trailing two axes (member, day) and sort ascending."""
    return np.sort(arr.reshape(arr.shape[:-2] + (-1,)), axis=-1)


def rank_sort(da):
    """Per season, pool members and days into one sorted `rank` axis."""
    da = to_year_day(da)
    n_sample = da.sizes['member'] * da.sizes['day']
    ranked = xr.apply_ufunc(
        sort_pooled,
        da,
        input_core_dims=[['member', 'day']],
        output_core_dims=[['rank']],
    )
    return ranked.assign_coords(
        quantile=('rank', (np.arange(n_sample) + 0.5) / n_sample)
    )


# ------------------------------------------------------------ trimming curves

def trim_curve(quantile_diff, from_hot_end):
    """Mean hist-histnat difference after removing n_removed days from one end.

    from_hot_end=True removes the hottest days first, False the coldest.
    """
    diff = quantile_diff.transpose(..., 'rank')
    values = diff.values
    if from_hot_end:
        pass  # cumulate from the cold end, i.e. hottest are dropped first
    else:
        values = values[..., ::-1]

    n_rank = diff.sizes['rank']
    retained = np.arange(1, n_rank + 1)
    means = np.cumsum(values, axis=-1)[..., ::-1] / retained[::-1]

    return xr.DataArray(
        means,
        dims=diff.dims[:-1] + ('n_removed',),
        coords={dim: diff.coords[dim] for dim in diff.dims[:-1]},
    ).assign_coords(n_removed=('n_removed', np.arange(n_rank)))


def first_crossing(curve):
    """Smallest n_removed at which the trimmed mean flips sign. NaN if never."""
    usable = curve.n_removed <= curve.sizes['n_removed'] - MIN_RETAINED
    reference = np.sign(curve.isel(n_removed=0))
    crossed = (np.sign(curve) != reference) & usable
    return xr.where(
        crossed.any('n_removed'),
        crossed.argmax('n_removed'),
        np.nan,
    )


# --------------------------------------------------------------- decomposition

def rank_decomposition(hist_season, histnat_season):
    hist_q = rank_sort(hist_season)
    histnat_q = rank_sort(histnat_season)
    quantile_diff = hist_q - histnat_q

    trim_hot = trim_curve(quantile_diff, from_hot_end=True)
    trim_cold = trim_curve(quantile_diff, from_hot_end=False)

    hist_anom = hist_q - hist_q.mean('rank')
    histnat_anom = histnat_q - histnat_q.mean('rank')
    qq_slope = (histnat_anom * hist_anom).sum('rank') / (
        histnat_anom ** 2
    ).sum('rank')

    results = xr.Dataset({
        'hist_q': hist_q,
        'histnat_q': histnat_q,
        'quantile_diff': quantile_diff,
        'mean_diff': quantile_diff.mean('rank'),
        'trim_hot': trim_hot,
        'trim_cold': trim_cold,
        'k_star_hot': first_crossing(trim_hot),
        'k_star_cold': first_crossing(trim_cold),
        'qq_slope': qq_slope,
        'qq_intercept': hist_q.mean('rank') - qq_slope * histnat_q.mean('rank'),
    })

    quantile = quantile_diff.quantile
    for name, band in BANDS.items():
        within = (quantile >= band['lower']) & (quantile <= band['upper'])
        results[f'diff_{name}'] = quantile_diff.where(within).mean('rank')

    return results

```

```python

# ------------------------------------------------------------------- plotting

def plot_season(results, season_year, axes=None):
    """Trimming curve and rank-difference curve for one season."""
    one = results.sel(season_year=season_year)
    if axes is None:
        _, axes = plt.subplots(1, 2, figsize=(11, 4))

    curves = {
        'hottest removed': {'data': one.trim_hot, 'colour': 'firebrick'},
        'coldest removed': {'data': one.trim_cold, 'colour': 'steelblue'},
    }
    for label, curve in curves.items():
        axes[0].plot(
            curve['data'].n_removed,
            curve['data'],
            color=curve['colour'],
            label=label,
        )
    axes[0].axhline(0, color='k', lw=0.6)
    axes[0].axhline(float(one.mean_diff), color='k', lw=0.6, ls='--')
    axes[0].set_xlabel('days removed')
    axes[0].set_ylabel('trimmed mean difference (K)')
    axes[0].set_title(f'DJF {int(season_year)}')
    axes[0].legend(frameon=False)

    axes[1].plot(one.quantile, one.quantile_diff, color='k', lw=0.8)
    axes[1].axhline(float(one.mean_diff), color='k', lw=0.6, ls='--')
    axes[1].set_xlabel('quantile')
    axes[1].set_ylabel('hist - histnat (K)')
    return axes


def plot_record(results, ax=None):
    """All seasons at once: quantile on x, season on y, difference in colour."""
    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))
    results.quantile_diff.plot(
        x='quantile', y='season_year', center=0, robust=True, ax=ax
    )
    ax.set_xlabel('quantile')
    ax.set_ylabel('season year')
    return ax


def plot_quantile_functions(results, season_year, ax=None):
    """Absolute quantile functions, for checking a melt ceiling near 273.15 K."""
    one = results.sel(season_year=season_year)
    if ax is None:
        _, ax = plt.subplots(figsize=(6, 4))

    experiments = {
        'hist': {'data': one.hist_q, 'colour': 'firebrick'},
        'histnat': {'data': one.histnat_q, 'colour': 'steelblue'},
    }
    for label, experiment in experiments.items():
        ax.plot(
            one.quantile,
            experiment['data'],
            color=experiment['colour'],
            label=label,
        )
    ax.axhline(273.15, color='k', lw=0.6, ls=':')
    ax.set_xlabel('quantile')
    ax.set_ylabel('temperature (K)')
    ax.legend(frameon=False)
    return ax
```

# Probability Ratio

There are many variations of how the probability ratio can be calculated.
The choices fall into three stages, and they are largely independent.

---

**1. Threshold definition**

How the extreme is defined, from the `hist-nat` distribution.

- **Rolling day of year** — *preferential*. Currently a $\pm 21$ day window.
  - *Weights*: currently uniform; could taper away from the central day.
  - *Window width*: an individual day works in the pooled case only, where the
    sample size is large enough.
- **Seasonal** — *not preferred*, but still tested.
---

**2. Counterfactual probability**

What the `hist` exceedance rate is compared against.

- **Fixed value** — a single number from the pooled `hist-nat`.
- **Per year** — estimated against the concurrent `hist-nat` year, so natural
  variability in PR is retained.

---

**3. Ensemble aggregation**

Applies at both stages above, which is why it is listed separately.

- **Per member** — PR computed for each member, giving a spread.
- **Pooled** — exceedances summed across members before taking the ratio.
  Allows really extreme percentiles (e.g. `99.9`), but loses the uncertainty
  estimate.

---

**4. Estimation window**

Over how many years the exceedance rate is counted.

- **Per year** — noisy, but no end point issues, and maybe not as good for larger percentiles (less sample size)
- **Rolling $\pm 10$ years** — 21 years centred on the target year.

---

```python
def exceedance_flags(da, q_da):
    """Flag days exceeding a day-of-year quantile threshold.
    Selects the threshold matching each timestep's calendar day and
    compares elementwise, returning a boolean on the original time axis
    rather than an aggregate, so the caller chooses how to bin it. Any
    coordinates the threshold selection carries in, such as a scalar
    ``height`` or ``quantile`` and the broadcast ``dayofyear``, are
    dropped, since they are not dims of the result and interfere with
    later stacking. Missing values in ``da`` compare False and so read as
    non-exceedances rather than as gaps; count valid days separately if
    that distinction matters. ``q_da`` must be on the same calendar as
    ``da``, since the match is on day-of-year number alone.
    Parameters
    ----------
    da : xr.DataArray
        Data with a datetime ``time`` dim.
    q_da : xr.DataArray
        Thresholds with a ``dayofyear`` dim covering the calendar.
    Returns
    -------
    xr.DataArray
        Boolean, same shape as ``da``, True where the threshold is
        exceeded.
    """
    q_doy = q_da.sel(dayofyear=da.time.dt.dayofyear)
    exceed = da > q_doy
    exceed = exceed.drop_vars([c for c in exceed.coords if c not in exceed.dims])
    return exceed


def doy_threshold_seasonal_probability(ensemble_ds, q_doy):
    """Compute the seasonal probability of threshold exceedance for one ensemble.
    Counts days exceeding the day-of-year quantile threshold, aggregates
    those counts into DJF/MAM/JJA/SON bins, and divides by the number of
    valid days in each bin to give an exceedance probability rather than a
    raw count. Bins are ``QS-DEC``, so DJF at year Y spans Dec of Y plus
    Jan and Feb of Y+1. Numerator and denominator are resampled with
    identical bin edges, so the two share a time axis exactly and the
    division needs no alignment. The first and last bins of the record are
    partial and their probabilities are estimated from fewer days than the
    rest; mask on ``valid`` if that matters for the downstream statistics.
    Parameters
    ----------
    ensemble_ds : xr.DataArray
        Daily data with a ``time`` dim on a ``360_day`` calendar, matching
        the input to ``exceedance_flags``.
    q_doy : xr.DataArray
        Day-of-year quantile thresholds, passed straight to
        ``exceedance_flags``.
    Returns
    -------
    xr.Dataset
        ``year`` and ``season`` dims, with variables ``exceed`` (days over
        the threshold), ``valid`` (non-NaN days contributing to the bin)
        and ``p`` (the exceedance probability).
    """
    # Exceedance flags on the original time axis.
    exceed_doy = exceedance_flags(ensemble_ds, q_doy)
    # Same binning for both. count() skips NaN, so valid is days with
    # data rather than days in the calendar quarter.
    exceed = exceed_doy.resample(time='QS-DEC').sum()
    valid = ensemble_ds.resample(time='QS-DEC').count()
    # Divide while both are still on the shared time axis, then split
    # once. Splitting first would mean trusting two independently
    # reconstructed (year, season) grids to line up.
    probability = exceed / valid
    exceed = split_time(exceed, ('year', 'season'), 'time')
    valid = split_time(valid, ('year', 'season'), 'time')
    probability = split_time(probability, ('year', 'season'), 'time')
    exceed.name = 'exceed'
    valid.name = 'valid'
    probability.name = 'p'
    merged = xr.merge([exceed, valid, probability])
    return merged
```

```python
def rolling_windows(season_ds, windows=(3, 5, 11, 21), center=True):
    """Pool seasonal exceedance statistics over a range of window lengths.
    Sums ``exceed`` and ``valid`` over centred rolling windows of whole
    years and recomputes ``p`` from the pooled totals, rather than
    averaging the annual probabilities, so years with fewer valid days
    are weighted accordingly. Incomplete windows at the ends come back
    as NaN.
    Parameters
    ----------
    season_ds : xr.Dataset
        Output of ``seasonal_exceedance_probability``.
    windows : sequence of int
        Window lengths in years, stacked along a new ``window`` dim.
    center : bool, default True
        Label each window at its midpoint rather than its right edge.
    Returns
    -------
    xr.Dataset
        Same as ``season_ds`` with an added ``window`` dim.
    """
    to_concat = [season_ds.assign_coords(window=1)]
    for w in windows:
        pooled = season_ds[['exceed', 'valid']].rolling(year=w, center=center).sum()
        pooled['p'] = pooled.exceed / pooled.valid
        pooled = pooled.assign_coords(window=w)
        to_concat.append(pooled)
    out = xr.concat(to_concat, dim='window')
    return out
```

```python
def subsample_windows(count_ds, window=11, size=10, n_runs=100):
    """Bootstrap the windowed exceedance probability over random member draws.

    Rolls all members once, then sums the drawn subset, exploiting the counts
    being additive. Draws are with replacement. Runs are stacked on a ``sample``
    dim, not ``member``, so the result does not align against a per-member
    counterfactual. Seeded internally, so successive calls give different draws.
    The returned counts are pooled over the drawn members, so they are larger
    than the per-member counts going in and double-count repeated draws.

    Args:
        count_ds (xr.Dataset): Counts with ``exceed`` and ``valid``, on ``year``
            and ``member`` dims. Any other dims broadcast.
        window (int): Window length in years.
        size (int): Members drawn per run.
        n_runs (int): Number of runs.

    Returns:
        xr.Dataset: ``exceed``, ``valid`` and ``p``, with an added ``sample`` dim.
    """
    rng = np.random.default_rng()

    exceed_rolled = count_ds.exceed.rolling(year=window, center=True).sum()
    valid_rolled = count_ds.valid.rolling(year=window, center=True).sum()

    runs = []
    for i in range(n_runs):
        draw = rng.choice(count_ds.sizes['member'], size=size, replace=True)
        exceed = exceed_rolled.isel(member=draw).sum(dim='member')
        valid = valid_rolled.isel(member=draw).sum(dim='member')
        runs.append(
            xr.Dataset({'exceed': exceed, 'valid': valid, 'p': exceed / valid}).assign_coords(sample=i)
        )
    return xr.concat(runs, dim='sample')
```

## Part 1: Climatology Defining


All climatologies are day-of-year specific. Two choices combine: how ensemble
members are treated, and whether a day window is used.

**Single day** — the *q*th percentile computed over years only, for each calendar day:

- **a. Individual members** — percentile across time, separately per member.
- **b. Pooled members** — all members combined. As these are initial-condition
  ensemble members, pooling gives the best estimate of the higher percentiles.

**Windowed** — as above, but the sample includes the 10 days either side of the
central day (21 days total):

- **c. Individual members**
- **d. Pooled members**

Note: Option 2. needs to be assessed with how sensitive it is to the number of ensemble members. This is as other models will have less ensemble members.

```python
def rolling_quantile_exceedance_dayofyear(da, q_da, resample_freq='QS-DEC', year_window: int = 21
                                         ) -> xr.Dataset:
    """Rolling exceedance rate relative to a seasonal quantile threshold.

    Counts timesteps in each resampled season that exceed the matching seasonal
    threshold, then accumulates those counts (and the number of valid timesteps)
    over a centred rolling window to give an exceedance probability.

    Args:
        da (xr.DataArray): Input data with a 'time' coordinate.
        q_da (xr.DataArray): Thresholds with a 'season' dim, broadcast against the seasonal groups of `da`.
        resample_freq (str): Frequency used to bin timesteps into seasons.
        year_window (int): Width of the centred rolling window, in years.

    Returns:
        xr.Dataset: 'pr' (exceedance probability), 'exceed' and 'total' (rolling counts).
            `resample_freq` and `year_window` are recorded in the attrs.
    """
    # per-season counts of exceedances and of valid timesteps
    n_exceed_da = (da > q_da).resample(time=resample_freq).sum()
    n_total_da = da.resample(time=resample_freq).count()

    # move onto a (year, season) index, then accumulate over the centred window
    roll_exceed_da = year_season(n_exceed_da).rolling(year=year_window, center=True).sum()
    roll_total_da = year_season(n_total_da).rolling(year=year_window, center=True).sum()

    # fraction of timesteps in the window above the threshold
    p_da = roll_exceed_da / roll_total_da
    
    roll_exceed_da.name = 'exceed'
    roll_total_da.name = 'total'
    p_da.name = 'p'

    ds = xr.merge([p_da, roll_total_da, roll_exceed_da])
    ds.attrs['resample_freq'] = resample_freq
    ds.attrs['year_window'] = year_window
    return ds
```

```python
def rolling_windows(season_ds, windows=(3, 5, 11, 21), center=True):
    """Pool seasonal exceedance statistics over a range of window lengths.
    Sums ``exceed`` and ``valid`` over centred rolling windows of whole
    years and recomputes ``p`` from the pooled totals, rather than
    averaging the annual probabilities, so years with fewer valid days
    are weighted accordingly. Incomplete windows at the ends come back
    as NaN.
    Parameters
    ----------
    season_ds : xr.Dataset
        Output of ``seasonal_exceedance_probability``.
    windows : sequence of int
        Window lengths in years, stacked along a new ``window`` dim.
    center : bool, default True
        Label each window at its midpoint rather than its right edge.
    Returns
    -------
    xr.Dataset
        Same as ``season_ds`` with an added ``window`` dim.
    """
    to_concat = []
    for w in windows:
        pooled = season_ds[['exceed', 'valid']].rolling(year=w, center=center).sum()
        pooled['p'] = pooled.exceed / pooled.valid
        pooled = pooled.assign_coords(window=w)
        to_concat.append(pooled)
    out = xr.concat(to_concat, dim='window')
    return out
```

# Rank Histograms

Using the method from Suarez‑Gutierrez et al., 2021 that uses rank histograms to analyse if both the signal and the nosie are correct in the model. This can get you if the forced resonse is under or overestimated, and if the variable is under or overestimated

Method
1. Convert to seasonal mean.
2. Convert to anomalies: subtract the mean of all data
3. Rank the data per time step against the ensemble members
4. Do the same for each ensemble member against all the other members
5. Calculate the 5th and 95th percentiles for each bin from the ensemble member ranks
6. Compare the rank histogram for the observations against the 5th and 9th percentiles

```python
def grid_slope(da, dim='time'):
    """Per-cell OLS slope of y = a + b*x.

    Closed-form normal-equation solution:

        b   = Sxy / Sxx
        Sxx = sum((x - xbar)^2)          spread of the time axis
        Sxy = sum((x - xbar)(y - ybar))  covariation of y with time

    Vectorised over all non-`dim` dimensions, so no per-cell loop.
    Time is converted to float years, since cftime objects don't support
    the arithmetic. Assumes one value per year — sub-annual data collapses
    Sxx, as every point within a year shares the same x.

    Parameters
    ----------
    da : xr.DataArray
        Input with cftime/datetime `dim`; other dims treated as cells.
    dim : str
        Dimension to regress along.

    Returns
    -------
    xr.DataArray
        Slope of the data.
    """
    x = xr.DataArray(np.arange(da.sizes[dim], dtype=float), dims=dim)

    dx = x - x.mean(dim)
    slope = (dx * (da - da.mean(dim))).sum(dim) / (dx ** 2).sum(dim)
    # slope.attrs['units'] = f"{da.attrs.get('units', '')} yr-1".strip()
    return slope
```

```python
import numpy as np
import xarray as xr
from scipy.stats import norm


def mk_S(y, i, j):
    """Mann-Kendall S statistic along the last axis.

    S = sum_{k<l} sign(y_l - y_k), as a broadcast pairwise difference
    over precomputed upper-triangle index pairs.

    Parameters
    ----------
    y : np.ndarray
        Shape (..., n); leading axes treated as independent cells.
    i, j : np.ndarray
        Upper-triangle index pairs from np.triu_indices(n, k=1).

    Returns
    -------
    np.ndarray
        S with shape y.shape[:-1].
    """
    return np.sign(y[..., j] - y[..., i]).sum(axis=-1)


def grid_mk(da, dim='time', alpha=0.05):
    """Per-cell Mann-Kendall test, lazy over dask.

        Var(S) = n(n-1)(2n+5) / 18          no-ties form
        Z      = (S -/+ 1) / sqrt(Var)      continuity-corrected, 0 if S = 0
        p      = 2 * P(N(0,1) > |Z|)

    apply_ufunc moves `dim` to the last axis, which is what mk_S expects.

    Parameters
    ----------
    da : xr.DataArray
        Input; `dim` is the time axis, other dims treated as cells.
    dim : str
        Dimension to test along.
    alpha : float
        Threshold for the significance mask.

    Returns
    -------
    xr.Dataset
        S, Z, p (two-sided), sig (bool, p < alpha).
    """
    n = da.sizes[dim]
    i, j = np.triu_indices(n, k=1)
    var = n * (n - 1) * (2 * n + 5) / 18.0

    S = xr.apply_ufunc(
        lambda y: mk_S(y, i, j),
        da,
        input_core_dims=[[dim]],
        dask='parallelized', output_dtypes=[float]
        )

    Z = xr.where(S > 0, (S - 1) / np.sqrt(var),
        xr.where(S < 0, (S + 1) / np.sqrt(var), 0.0))
    p = xr.apply_ufunc(lambda z: 2 * norm.sf(np.abs(z)), Z,
                       dask='parallelized', output_dtypes=[float])

    return xr.Dataset({'S': S, 'Z': Z, 'p': p, 'sig': p < alpha})
```

```python
def grid_intercept(da, slope, dim='time'):
    """Per-cell OLS intercept of y = a + b*x, given b.

        a = ybar - b * xbar

    Uses the same x as `grid_slope`: 0, 1, ..., n-1 along `dim`. The
    intercept is therefore the fitted value at the *first* time step,
    not at any calendar epoch.

    Parameters
    ----------
    da : xr.DataArray
        Input with cftime/datetime `dim`; other dims treated as cells.
    slope : xr.DataArray
        Slope from `grid_slope(da, dim)`, broadcastable against `da`
        over the non-`dim` dims.
    dim : str
        Dimension the regression was taken along.

    Returns
    -------
    xr.DataArray
        Intercept of the data.
    """
    x = xr.DataArray(np.arange(da.sizes[dim], dtype=float), dims=dim)
    return da.mean(dim) - slope * x.mean(dim)
```

```python
import matplotlib.pyplot as plt


def make_axes(fig, npanels, layout="col", height_ratios=(5, 1)):
    axes, tab_axes = [], []
    if layout == "col":
        gs = fig.add_gridspec(2, npanels, height_ratios=height_ratios)
        slots = [(gs[0, i], gs[1, i]) for i in range(npanels)]
    elif layout == "row":
        gs = fig.add_gridspec(2 * npanels, 1,
                              height_ratios=list(height_ratios) * npanels)
        slots = [(gs[2 * i, 0], gs[2 * i + 1, 0]) for i in range(npanels)]
    else:
        raise ValueError("layout must be 'col' or 'row'")

    for plot_slot, tab_slot in slots:
        axes.append(fig.add_subplot(plot_slot, sharex=axes[0] if axes else None))
        tab_axes.append(fig.add_subplot(tab_slot))
    return axes, tab_axes


def plot_signal_sn(points, left_key, right_key, table_keys,
                   layout="col", figsize=None, height_ratios=(5, 1),
                   left_color="tab:blue", right_color="tab:red",
                   left_label="Temperature (K)"):
    n = len(points)
    if figsize is None:
        figsize = (6 * n, 5) if layout == "col" else (7, 3.5 * n)

    fig = plt.figure(figsize=figsize, constrained_layout=True)
    axes, tab_axes = make_axes(fig, n, layout, height_ratios)

    for ax, ax_tab, p in zip(axes, tab_axes, points):
        draw_panel(ax, p["label"], p["pt"], p["data"], left_key, right_key,
                   left_color=left_color, right_color=right_color,
                   left_label=left_label)
        draw_table(ax_tab, p["data"], table_keys)

    return fig, axes
```

```python
def draw_panel(ax, label, pt, data, left_key, right_key,
               left_color="tab:blue", right_color="tab:red",
               left_label="Temperature (K)"):
    left, right = data[left_key], data[right_key]

    l1, = ax.plot(left.time, left, color=left_color, lw=1.2,
                  label=f"{left_key}")
    ax.set_ylabel(left_label, color=left_color)
    ax.tick_params(axis="y", labelcolor=left_color, length=3)
    ax.tick_params(axis="x", length=3)
    ax.set_title(f"{label}: {pt['lat']}°, {pt['lon']}°", fontsize=11, pad=8)
    ax.grid(alpha=0.25, lw=0.6)
    ax.set_axisbelow(True)

    ax2 = ax.twinx()
    l2, = ax2.plot(right.time, right, color=right_color, lw=1.8,
                   label=f"{right_key} ({endpoint(right):.2f})")
    ax2.set_ylabel(right_key, color=right_color)
    ax2.tick_params(axis="y", labelcolor=right_color, length=3)
    ax2.axhline(0, color="0.5", lw=0.8, ls=":")

    ax.spines["top"].set_visible(False)
    ax2.spines["top"].set_visible(False)
    ax.legend(handles=[l1, l2], loc="upper left", frameon=False, fontsize=12)
    return ax2
```

```python
def draw_table(ax, data, table_keys):
    ax.axis("off")
    tab = ax.table(cellText=[[f"{endpoint(data[k]):.2f}" for k in table_keys]],
                   colLabels=list(table_keys),
                   cellLoc="center", loc="center")
    tab.auto_set_font_size(False)
    tab.set_fontsize(9)
    tab.scale(1, 1.3)
    for (r, c), cell in tab.get_celld().items():
        cell.set_edgecolor("0.8")
        if r == 0:
            cell.set_text_props(weight="bold")
    return tab
```

```python
def to_year(da):
    return da.assign_coords(time=da.time.dt.year) if "time" in da.dims else da


def build_points(stations, arrays):
    return [{"label": label,
             "pt": pt,
             "data": {name: to_year(da).sel(lat=pt["lat"], lon=pt["lon"],
                                            method="nearest").compute()
                      for name, da in arrays.items()}}
            for label, pt in stations.items()]
```

```python
def make_axes(fig, npanels, n_series_rows=2, layout="col",
              height_ratios=(3, 3, 2, 1)):
    nrows = n_series_rows + 2
    series_axes, sn_axes, tab_axes = [], [], []
    if layout == "col":
        gs = fig.add_gridspec(nrows, npanels, height_ratios=height_ratios)
        slots = [[gs[r, i] for r in range(nrows)] for i in range(npanels)]
    elif layout == "row":
        gs = fig.add_gridspec(nrows * npanels, 1,
                              height_ratios=list(height_ratios) * npanels)
        slots = [[gs[nrows * i + r, 0] for r in range(nrows)]
                 for i in range(npanels)]
    else:
        raise ValueError("layout must be 'col' or 'row'")

    for col_slots in slots:
        shared_x = series_axes[0][0] if series_axes else None
        col = []
        for r in range(n_series_rows):
            ax = fig.add_subplot(col_slots[r], sharex=shared_x,
                                 sharey=col[0] if col else None)
            shared_x = shared_x or ax
            col.append(ax)
        series_axes.append(col)
        sn_axes.append(fig.add_subplot(col_slots[-2], sharex=col[0]))
        tab_axes.append(fig.add_subplot(col_slots[-1]))
    return series_axes, sn_axes, tab_axes

def draw_compare_panel(ax_list, ax_sn, label, pt, data, series_rows, sn_keys,
                       series_label="Anomaly (K)", sn_label="S/N ratio",
                       row_labels=None, show_row_labels=True):
    for i, (ax, row) in enumerate(zip(ax_list, series_rows)):
        for key, style in row.items():
            data[key].plot(ax=ax, **style)
        ax.axhline(0, color="0.75", lw=0.6, zorder=0)
        ax.set_ylabel(series_label)
        ax.set_xlabel("")
        ax.set_title("")
        ax.tick_params(labelbottom=False)
        ax.legend(fontsize=8, loc="upper left", framealpha=0.9, ncol=2)
        if row_labels and show_row_labels:
            ax.text(-0.22, 0.5, row_labels[i], transform=ax.transAxes,
                    rotation=90, va="center", ha="center",
                    fontsize=11, fontweight="bold")

    for key, style in sn_keys.items():
        data[key].plot(ax=ax_sn, **style)
    for y in (-2, -1, 1, 2):
        ax_sn.axhline(y, color="0.75", lw=0.6, ls=":", zorder=0)
    ax_sn.axhline(0, color="0.75", lw=0.6, zorder=0)
    ax_sn.set_ylabel(sn_label)
    ax_sn.set_xlabel("Year")
    ax_sn.set_title("")
    ax_sn.legend(fontsize=8, loc="upper left", framealpha=0.9, ncol=2)
    if row_labels and show_row_labels and len(row_labels) > len(series_rows):
        ax_sn.text(-0.22, 0.5, row_labels[-1], transform=ax_sn.transAxes,
                   rotation=90, va="center", ha="center",
                   fontsize=11, fontweight="bold")

    ax_list[0].set_title(f"{label}   {pt['lat']:.1f}°, {pt['lon']:.1f}°",
                         loc="left", fontsize=11, fontweight="bold")


def plot_compare(points, series_rows, sn_keys, table_keys,
                 layout="col", figsize=None, height_ratios=(3, 3, 2, 1),
                 row_labels=None, **kwargs):
    n = len(points)
    if figsize is None:
        figsize = (5.5 * n, 8.5) if layout == "col" else (7, 6 * n)
    fig = plt.figure(figsize=figsize, constrained_layout=True)
    series_axes, sn_axes, tab_axes = make_axes(fig, n, len(series_rows),
                                               layout, height_ratios)
    for i, (ax_list, ax_sn, ax_tab, p) in enumerate(
            zip(series_axes, sn_axes, tab_axes, points)):
        draw_compare_panel(ax_list, ax_sn, p["label"], p["pt"], p["data"],
                           series_rows, sn_keys, row_labels=row_labels,
                           show_row_labels=(i == 0 or layout == "row"), **kwargs)
        draw_table(ax_tab, p["data"], table_keys)
    return fig, series_axes, sn_axes
```

```python
import numpy as np


def sorted_percentile(sorted_vals, percentile, n_valid=None):
    """Extract a percentile from a pre-sorted 1D array. O(1).

    Linear interpolation between order statistics, matching np.percentile's
    default (method='linear').

    sorted_vals: sorted ascending, NaNs at the end.
    percentile: in [0, 100].
    n_valid: number of non-NaN values. If None, uses the whole array.
    """
    if np.all(np.isnan(sorted_vals)): return np.nan
        
    if n_valid is None:
        n_valid = len(sorted_vals)
    if n_valid == 0:
        return np.nan

    # Fractional index into the valid block: 0 -> index 0, 100 -> n_valid-1.
    pos = (n_valid - 1) * percentile / 100.0
    below = int(np.floor(pos))
    above = min(below + 1, n_valid - 1)  # clip so 100 doesn't run past the end

    return sorted_vals[below] + (pos - below) * (
        sorted_vals[above] - sorted_vals[below]
    )
```

```python
def exceedances(sorted_vals, threshold, n_valid=None):
    """Values strictly above a threshold, from pre-sorted data.

    sorted_vals: sorted ascending, NaNs at the end.
    threshold: exceedances are strictly greater than this.
    n_valid: number of non-NaN values. If None, uses the whole array.
    """
    if n_valid is None:
        n_valid = len(sorted_vals)

    # Binary search for first index above threshold. Slicing to n_valid
    # excludes NaNs, which sort as greater than everything.
    start = np.searchsorted(sorted_vals[:n_valid], threshold, side='right')
    return sorted_vals[start:n_valid]
```

```python
def threshold_exceedances_sorted(sorted_vals, threshold, n_valid=None):
    """Values strictly above a threshold, from pre-sorted data.

    sorted_vals: sorted ascending, NaNs at the end.
    threshold: exceedances are strictly greater than this.
    n_valid: number of non-NaN values. If None, uses the whole array.
    """
    if n_valid is None:
        n_valid = len(sorted_vals)

    # Binary search for first index above threshold. Slicing to n_valid
    # excludes NaNs, which sort as greater than everything.
    start = np.searchsorted(sorted_vals[:n_valid], threshold, side='right')
    return sorted_vals[start:n_valid]
```
